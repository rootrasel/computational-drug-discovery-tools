<h4 style="margin: 0 0 10px 0; color: #1e1b4b; font-size: 16px; font-weight: 700;">
Why Molecular Scaffold Splitting is Essential in Drug Discovery
</h4>
<p style="font-size: 14px; color: #334155; margin-bottom: 12px;">
In traditional machine learning, data is often partitioned randomly. However, random splitting in cheminformatics causes <b>severe structural data leakage</b>: structurally related molecules (analogs sharing identical core scaffolds) end up in both the training and test sets. This produces artificially inflated test metrics ($R^2 > 0.85$) that collapse when deployed on novel chemotypes.
</p>
<div style="background: #ffffff; border: 1px solid #c7d2fe; border-radius: 10px; padding: 14px; margin-bottom: 14px;">
<div style="font-weight: 700; color: #4338ca; font-size: 14px; margin-bottom: 6px;">Bemis-Murcko Scaffold Splitting:</div>
<div style="font-size: 13.5px; color: #334155;">
DeepChem's <code>ScaffoldSplitter</code> isolates the two-dimensional ring systems and linkers (Bemis-Murcko scaffolds) of each molecule. Molecules with the same core scaffold are strictly assigned to either the training set, validation set, or test set. This realistically tests the model's ability to extrapolate to <b>novel chemical scaffolds</b>, mirroring prospective lead discovery.
</div>
</div>
<p style="font-size: 13.5px; color: #475569; margin: 0;">
We load Delaney with <code>featurizer='ECFP'</code> (1024-bit Morgan Fingerprints, radius 2) and <code>splitter='scaffold'</code>.
</p>
</div>


```python
# Load Delaney (ESOL) with ECFP Featurizer and Scaffold Splitter
tasks, datasets, transformers = dc.molnet.load_delaney(
    featurizer="ECFP",
    splitter="scaffold",
    transformers=[]
)

train_dataset, valid_dataset, test_dataset = datasets

print("=" * 70)
print("DELANEY (ESOL) BENCHMARK DATASET LOADED")
print("=" * 70)
print(f"Target Task Name     : {tasks[0]}")
print(f"Transformers Applied : {transformers} (Unscaled raw units)")
print(f"Training Samples     : {len(train_dataset)} ({len(train_dataset)/1128*100:.1f}%)")
print(f"Validation Samples   : {len(valid_dataset)} ({len(valid_dataset)/1128*100:.1f}%)")
print(f"Test Samples         : {len(test_dataset)} ({len(test_dataset)/1128*100:.1f}%)")
print(f"Total Compounds      : {len(train_dataset) + len(valid_dataset) + len(test_dataset)}")
print("-" * 70)
print(f"Feature Matrix Shape (X) : {train_dataset.X.shape} (N samples, 1024-bit ECFP)")
print(f"Target Vector Shape (y)  : {train_dataset.y.shape} (N samples, 1 continuous target)")
print(f"Sample Weights Shape (w) : {train_dataset.w.shape} (Sample validity weights)")
print(f"Sample IDs (SMILES) Shape: {train_dataset.ids.shape}")
print("-" * 70)
print(f"Sample IDs (First 3 SMILES): {train_dataset.ids[:3]}")
```

Pre-Flight Data Quality Checklist for Drug Discovery
</h4>
<p style="font-size: 14px; color: #334155; margin-bottom: 12px;">
Before training any predictive model, we must execute thorough quality assurance checks:
</p>
<ul style="font-size: 13.5px; color: #334155; margin-left: 20px; line-height: 1.7;">
<li><b>Target Definition & Scale:</b> The target is $\\log S$ (base-10 logarithm of molar solubility, $\text{mol/L}$).
<ul>
<li>$\\log S > -2$: Highly soluble (freely dissolving compounds).</li>
<li>$-4 < \\log S \le -2$: Moderately soluble (ideal for oral bioavailability).</li>
<li>$\\log S \le -4$: Poorly soluble (high risk of formulation failure and poor absorption).</li>
<li>$\\log S \le -6$: Practically insoluble.</li>
</ul>
</li>
<li><b>Completeness & Nulls:</b> Are there NaN, Inf, or empty values in the feature matrix $X$ or target vector $y$?</li>
<li><b>SMILES Parseability:</b> Can RDKit parse 100% of the SMILES identifiers without error?</li>
<li><b>Scaffold Leakage Verification:</b> Do any SMILES strings overlap between the training set and the test set?</li>
</ul>
</div>


```python
# 1. Check for Missing, Infinite, or Non-Finite Values
def audit_split(split_name, ds):
    nan_X = np.isnan(ds.X).sum()
    nan_y = np.isnan(ds.y).sum()
    inf_X = np.isinf(ds.X).sum()
    inf_y = np.isinf(ds.y).sum()
    
    # Validate SMILES parseability with RDKit
    valid_mols = sum(1 for smi in ds.ids if Chem.MolFromSmiles(smi) is not None)
    
    return {
        "Split": split_name,
        "Compounds": len(ds),
        "Feature Dim": ds.X.shape[1],
        "Missing X (NaN)": nan_X,
        "Missing y (NaN)": nan_y,
        "Non-Finite": inf_X + inf_y,
        "Valid RDKit SMILES": f"{valid_mols}/{len(ds)} (100%)",
        "Target Min (log S)": f"{float(np.min(ds.y)):.2f}",
        "Target Mean": f"{float(np.mean(ds.y)):.2f}",
        "Target Max (log S)": f"{float(np.max(ds.y)):.2f}"
    }

integrity_summary = pd.DataFrame([
    audit_split("Train", train_dataset),
    audit_split("Validation", valid_dataset),
    audit_split("Test", test_dataset)
])

# 2. Check for Structural Overlap / Chemical Leakage
train_smiles = set(train_dataset.ids)
valid_smiles = set(valid_dataset.ids)
test_smiles = set(test_dataset.ids)

leak_train_test = train_smiles.intersection(test_smiles)
leak_train_val = train_smiles.intersection(valid_smiles)

print("=" * 70)
print("DATASET INTEGRITY & CHEMICAL LEAKAGE AUDIT")
print("=" * 70)
print(f"Overlap between Train and Test : {len(leak_train_test)} compounds (Zero Leakage Verified)")
print(f"Overlap between Train and Valid: {len(leak_train_val)} compounds (Zero Leakage Verified)")
print("-" * 70)
display(integrity_summary)
```

<h4 style="margin: 0 0 10px 0; color: #1e1b4b; font-size: 16px; font-weight: 700;">
Why Every Professional Pipeline Needs a Naive Baseline
</h4>
<p style="font-size: 14px; color: #334155; margin-bottom: 12px;">
In machine learning research, a complex model's performance can only be judged relative to a <b>naive baseline</b>. A <code>DummyRegressor(strategy='mean')</code> makes a constant prediction equal to the training set mean $\bar{y}_{\text{train}}$ for every sample:
</p>
$$\hat{y}_{i} = \frac{1}{N_{\text{train}}} \sum_{j=1}^{N_{\text{train}}} y_j$$
<p style="font-size: 13.5px; color: #334155; margin-bottom: 12px;">
By mathematical definition:
</p>
<ul style="font-size: 13.5px; color: #334155; margin-left: 20px; line-height: 1.7;">
<li>On an identical distribution, the DummyRegressor achieves an $R^2 = 0.0$.</li>
<li>Under a <b>scaffold split</b> with distribution shift, the DummyRegressor typically achieves $R^2 < 0.0$ on the test set!</li>
<li>Any legitimate machine learning model must convincingly surpass the DummyRegressor's MAE and RMSE to prove that it is actually learning chemistry rather than merely guessing the dataset mean.</li>
</ul>
</div>


```python
# Initialize and fit naive baseline on Delaney training set
dummy_baseline = DummyRegressor(strategy="mean")
dummy_baseline.fit(train_dataset.X, train_dataset.y.ravel())

# Predict across all splits
pred_dummy_train = dummy_baseline.predict(train_dataset.X)
pred_dummy_val = dummy_baseline.predict(valid_dataset.X)
pred_dummy_test = dummy_baseline.predict(test_dataset.X)

def compute_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    return mae, rmse, r2

dummy_results = {
    "Split": ["Train", "Validation", "Test"],
    "MAE": [
        compute_metrics(train_dataset.y.ravel(), pred_dummy_train)[0],
        compute_metrics(valid_dataset.y.ravel(), pred_dummy_val)[0],
        compute_metrics(test_dataset.y.ravel(), pred_dummy_test)[0]
    ],
    "RMSE": [
        compute_metrics(train_dataset.y.ravel(), pred_dummy_train)[1],
        compute_metrics(valid_dataset.y.ravel(), pred_dummy_val)[1],
        compute_metrics(test_dataset.y.ravel(), pred_dummy_test)[1]
    ],
    "R² Score": [
        compute_metrics(train_dataset.y.ravel(), pred_dummy_train)[2],
        compute_metrics(valid_dataset.y.ravel(), pred_dummy_val)[2],
        compute_metrics(test_dataset.y.ravel(), pred_dummy_test)[2]
    ]
}

df_dummy = pd.DataFrame(dummy_results)
print("=" * 70)
print(f"REFERENCE NAIVE BASELINE (DummyRegressor predicting mean: {dummy_baseline.constant_[0][0]:.3f})")
print("=" * 70)
display(df_dummy.round(3))
```

<p style="font-size: 14px; color: #334155; margin-bottom: 12px;">
To answer Question 4 (<i>Can I train a simple baseline model and evaluate its performance?</i>), we now train a <b>Random Forest Regressor</b> (100 decision trees) on the 1024-bit ECFP fingerprints.
Random Forest is the premier classical benchmark in computational cheminformatics due to:

```python
# Train Random Forest Regressor on Delaney ECFP Fingerprints
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=RANDOM_SEED,
    n_jobs=-1
)

rf_model.fit(train_dataset.X, train_dataset.y.ravel())

# Predict across all splits
pred_rf_train = rf_model.predict(train_dataset.X)
pred_rf_val = rf_model.predict(valid_dataset.X)
pred_rf_test = rf_model.predict(test_dataset.X)

# Compute performance across all splits
rf_metrics = [
    compute_metrics(train_dataset.y.ravel(), pred_rf_train),
    compute_metrics(valid_dataset.y.ravel(), pred_rf_val),
    compute_metrics(test_dataset.y.ravel(), pred_rf_test)
]

comparison_table = pd.DataFrame({
    "Split": ["Train", "Validation", "Test"],
    "Dummy MAE": df_dummy["MAE"].values,
    "RF MAE": [m[0] for m in rf_metrics],
    "Dummy RMSE": df_dummy["RMSE"].values,
    "RF RMSE": [m[1] for m in rf_metrics],
    "Dummy R²": df_dummy["R² Score"].values,
    "RF R²": [m[2] for m in rf_metrics]
})

print("=" * 75)
print("BENCHMARK COMPARISON: DUMMY BASELINE vs. RANDOM FOREST REGRESSOR")
print("=" * 75)
display(comparison_table.round(3))
```

```python
# Generate Research-Grade Diagnostic Figures
fig, axes = plt.subplots(1, 2, figsize=(15, 6), dpi=120)

y_test_true = test_dataset.y.ravel()
y_test_pred = pred_rf_test
residuals = y_test_true - y_test_pred

# Plot 1: Parity Plot (Experimental vs Predicted)
axes[0].scatter(
    y_test_true,
    y_test_pred,
    color="#4f46e5",
    alpha=0.75,
    edgecolor="white",
    s=55,
    label=f"Test Molecules (N={len(y_test_true)})"
)

# Identity Line & Error Margin (+/- 1.0 log unit)
min_val = min(y_test_true.min(), y_test_pred.min()) - 0.5
max_val = max(y_test_true.max(), y_test_pred.max()) + 0.5
axes[0].plot([min_val, max_val], [min_val, max_val], color="#ef4444", linestyle="-", linewidth=1.8, label="Ideal Parity (y = x)")
axes[0].plot([min_val, max_val], [min_val + 1, max_val + 1], color="#94a3b8", linestyle="--", linewidth=1.2, label="+/- 1.0 Log Unit Error Band")
axes[0].plot([min_val, max_val], [min_val - 1, max_val - 1], color="#94a3b8", linestyle="--", linewidth=1.2)

test_r2 = r2_score(y_test_true, y_test_pred)
test_mae = mean_absolute_error(y_test_true, y_test_pred)
axes[0].set_title(f"Parity Plot: Test Set Performance (R² = {test_r2:.3f}, MAE = {test_mae:.3f})", pad=12)
axes[0].set_xlabel("Experimental Log Solubility (log mol/L)")
axes[0].set_ylabel("Predicted Log Solubility (log mol/L)")
axes[0].set_xlim(min_val, max_val)
axes[0].set_ylim(min_val, max_val)
axes[0].legend(frameon=True, facecolor="white", edgecolor="#cbd5e1", loc="upper left")

# Plot 2: Residuals vs. Predicted
axes[1].scatter(
    y_test_pred,
    residuals,
    color="#06b6d4",
    alpha=0.75,
    edgecolor="white",
    s=55
)
axes[1].axhline(0, color="#ef4444", linestyle="-", linewidth=1.8, label="Zero Error Line")
axes[1].axhline(1.0, color="#94a3b8", linestyle="--", linewidth=1.2)
axes[1].axhline(-1.0, color="#94a3b8", linestyle="--", linewidth=1.2, label="+/- 1.0 Residual Margin")
axes[1].set_title("Residual Analysis: Prediction Errors vs Fitted Values", pad=12)
axes[1].set_xlabel("Predicted Log Solubility (log mol/L)")
axes[1].set_ylabel("Residual Error (Actual - Predicted)")
axes[1].legend(frameon=True, facecolor="white", edgecolor="#cbd5e1", loc="upper left")

plt.tight_layout()
plt.show()
```

<h4 style="margin: 0 0 10px 0; color: #1e1b4b; font-size: 15px; font-weight: 700;">💡 Conceptual Review Questions</h4>
<details style="background: #f8fafc; border: 1px solid #cbd5e1; border-radius: 8px; padding: 12px 16px; margin-bottom: 10px; cursor: pointer;">
<summary style="font-weight: 600; color: #312e81; font-size: 14px;">1. Why does Scaffold Splitting yield lower test metrics than Random Splitting?</summary>
<p style="font-size: 13.5px; color: #334155; margin-top: 8px; line-height: 1.6;">
Random splitting places close structural analogs of training molecules into the test set, allowing the model to perform "memorization" or nearest-neighbor interpolation. Scaffold splitting forces the test set to contain entirely novel Bemis-Murcko core ring architectures that the model has never encountered, accurately measuring true generalization to uncharacterized chemotypes.
</p>
</details>

<details style="background: #f8fafc; border: 1px solid #cbd5e1; border-radius: 8px; padding: 12px 16px; margin-bottom: 10px; cursor: pointer;">
<summary style="font-weight: 600; color: #312e81; font-size: 14px;">2. What does a negative R² score for the DummyRegressor indicate?</summary>
<p style="font-size: 13.5px; color: #334155; margin-top: 8px; line-height: 1.6;">
A negative R² score means that predicting the training set mean performs worse than predicting the test set's own mean. In scaffold splits, the distribution of target properties in the test set may shift relative to the training set, causing a naive mean prediction to incur substantial residual sum of squares (SS_res > SS_tot).
</p>
</details>

<details style="background: #f8fafc; border: 1px solid #cbd5e1; border-radius: 8px; padding: 12px 16px; margin-bottom: 10px; cursor: pointer;">
<summary style="font-weight: 600; color: #312e81; font-size: 14px;">3. What is the physical significance of a +/- 1.0 log S error boundary?</summary>
<p style="font-size: 13.5px; color: #334155; margin-top: 8px; line-height: 1.6;">
Because log S is a base-10 logarithmic scale, a difference of 1.0 corresponds to a 10-fold change in molar solubility. In medicinal chemistry, inter-laboratory experimental solubility measurements typically exhibit an experimental error margin of 0.5 to 1.0 log units; thus, predictions within +/- 1.0 log unit are considered clinically and practically acceptable.
</p>
</details>

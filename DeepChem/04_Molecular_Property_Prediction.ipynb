```python
# 1. Environment and Logging Configuration
import os
import sys
import io
import logging
import warnings
import random

# Suppress TensorFlow C++ logging and oneDNN diagnostics
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
os.environ["TF_ENABLE_ONEDNN_OPTS"] = "0"
warnings.filterwarnings("ignore")
logging.getLogger("tensorflow").setLevel(logging.ERROR)
logging.getLogger("deepchem").setLevel(logging.ERROR)

# Suppress RDKit C++ deprecation warnings (e.g., MorganGenerator notices)
from rdkit import RDLogger
RDLogger.DisableLog("rdApp.*")

# 2. Quiet Import of DeepChem suppressing low-level C and stream stderr
old_stderr = sys.stderr
sys.stderr = io.StringIO()
devnull = os.open(os.devnull, os.O_WRONLY)
old_stderr_fd = os.dup(2)
try:
    os.dup2(devnull, 2)
    import tensorflow as tf
    tf.get_logger().setLevel("ERROR")
    import deepchem as dc
    from deepchem.models import GraphConvModel
finally:
    os.dup2(old_stderr_fd, 2)
    os.close(devnull)
    os.close(old_stderr_fd)
    sys.stderr = old_stderr

# 3. Scientific and Cheminformatics Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

import rdkit
from rdkit import Chem
from rdkit.Chem import Descriptors, Draw, AllChem

import sklearn
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 4. Set Global Seeds for Reproducibility
RANDOM_SEED = 42
os.environ["PYTHONHASHSEED"] = str(RANDOM_SEED)
random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

# Visual Styling Setup
plt.style.use("seaborn-v0_8-whitegrid" if "seaborn-v0_8-whitegrid" in plt.style.available else "default")
plt.rcParams.update({
    "font.family": "sans-serif",
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.titleweight": "bold",
    "axes.labelsize": 12,
    "axes.labelweight": "semibold",
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "figure.titlesize": 14,
    "figure.titleweight": "bold"
})

print(f"DeepChem Version  : {dc.__version__}")
print(f"RDKit Version     : {rdkit.__version__}")
print(f"Scikit-Learn Ver. : {sklearn.__version__}")
print(f"Pandas Version    : {pd.__version__}")
print(f"NumPy Version     : {np.__version__}")
print(f"Global Random Seed: {RANDOM_SEED} (Locked for Reproducibility)")
```

<h4 style="margin: 0 0 10px 0; color: #1e1b4b; font-size: 16px; font-weight: 700;">
Therapeutic and Physical Context of Aqueous Solubility
</h4>
<p style="font-size: 14px; color: #334155; margin-bottom: 12px;">
The Delaney benchmark (ESOL, <i>J. Chem. Inf. Comput. Sci.</i> 2004) consists of <b>1,128 diverse organic molecules</b> with experimentally measured 25°C aqueous solubility values expressed as $\log S$ (where $S$ is solubility in $\text{mols per litre}$).
</p>
<div style="background: #ffffff; border: 1px solid #c7d2fe; border-radius: 10px; padding: 14px; margin-bottom: 14px;">
<div style="font-weight: 700; color: #4338ca; font-size: 13.5px; margin-bottom: 6px;">The Medicinal Chemistry Solubility Scale:</div>
<ul style="font-size: 13px; color: #334155; margin: 0 0 0 18px; line-height: 1.7;">
<li><b>High Solubility ($\log S > -2$):</b> Compounds readily dissolve; no formulation dissolution barrier.</li>
<li><b>Moderate Solubility ($-4 < \log S \le -2$):</b> Standard drug-like range; suitable for oral absorption if permeability is favorable.</li>
<li><b>Low / Poor Solubility ($\log S \le -4$):</b> Significant risk of poor in vivo bioavailability; formulation requires nanosuspensions or lipid carriers.</li>
<li><b>Practically Insoluble ($\log S \le -6$):</b> Severe developability liability.</li>
</ul>
</div>

</h2>

<h4 style="margin: 0 0 10px 0; color: #1e1b4b; font-size: 16px; font-weight: 700;">
Synchronizing Scaffold Splits for Honest Model Comparison
</h4>
<p style="font-size: 14px; color: #334155; margin-bottom: 12px;">
To compare a traditional machine learning model against a Graph Neural Network with scientific rigor, <b>both models must be evaluated on the exact same molecular compounds in the test set</b>.
We load Delaney in two parallel tracks using DeepChem's deterministic <code>scaffold</code> splitter:
</p>
<ul style="font-size: 13.5px; color: #334155; margin-left: 20px; line-height: 1.7;">
<li><b>Track 1 (Tabular ECFP):</b> <code>featurizer='ECFP'</code> (1024-bit Morgan Fingerprints, radius 2) for our Random Forest regressor.</li>
<li><b>Track 2 (Graph ConvMol):</b> <code>featurizer='GraphConv'</code> (atomic feature vectors + adjacency graphs) for DeepChem's GraphConvModel.</li>
</ul>
</div>


```python
# Load Delaney under ECFP (Track 1) and GraphConv (Track 2) with scaffold splitting
print("=" * 70)
print("LOADING DELANEY WITH SYNCHRONIZED BEMIS-MURCKO SCAFFOLD SPLITS")
print("=" * 70)

# Track 1: ECFP Fingerprints
tasks_ecfp, datasets_ecfp, _ = dc.molnet.load_delaney(
    featurizer="ECFP",
    splitter="scaffold",
    transformers=[]
)
train_ecfp, val_ecfp, test_ecfp = datasets_ecfp

# Track 2: Graph Convolutions (ConvMol)
tasks_gc, datasets_gc, _ = dc.molnet.load_delaney(
    featurizer="GraphConv",
    splitter="scaffold",
    transformers=[]
)
train_gc, val_gc, test_gc = datasets_gc

# Verification of identical chemical assignments across tracks
assert list(test_ecfp.ids) == list(test_gc.ids), "CRITICAL: Scaffold test splits do not match!"
assert list(train_ecfp.ids) == list(train_gc.ids), "CRITICAL: Scaffold train splits do not match!"
assert list(val_ecfp.ids) == list(val_gc.ids), "CRITICAL: Scaffold validation splits do not match!"

print(">> SUCCESS: Verified 100% identical molecular IDs across both featurizations!")
print(f"Training Molecules   : {len(train_ecfp)} ({len(train_ecfp)/1128*100:.1f}%)")
print(f"Validation Molecules : {len(val_ecfp)} ({len(val_ecfp)/1128*100:.1f}%)")
print(f"Test Molecules       : {len(test_ecfp)} ({len(test_ecfp)/1128*100:.1f}%)")
print("-" * 70)
print(f"Track 1 (ECFP) Feature Matrix Shape     : {train_ecfp.X.shape}")
print(f"Track 2 (GraphConv) Sample Representation: {type(train_gc.X[0]).__name__} (ConvMol graph object)")
```

<p style="font-size: 14px; color: #334155; margin: 0;">
Before fitting predictive algorithms, we inspect the target property density across the train, validation, and test splits. Scaffold partitioning isolates structural families, which can introduce subtle target property shifts between training and test sets that models must navigate.
</p>
</div>


```python
# Compute statistical descriptors across splits
split_names = ["Train", "Validation", "Test"]
split_datasets = [train_ecfp, val_ecfp, test_ecfp]

eda_stats = []
for name, ds in zip(split_names, split_datasets):
    y = ds.y.ravel()
    eda_stats.append({
        "Split": name,
        "Compounds": len(y),
        "Mean (log S)": np.mean(y),
        "Std Dev": np.std(y),
        "Median": np.median(y),
        "Min (Least Soluble)": np.min(y),
        "Max (Most Soluble)": np.max(y)
    })

display(pd.DataFrame(eda_stats).set_index("Split").round(3))

# Quick KDE overlay plot
plt.figure(figsize=(9, 4), dpi=120)
x_grid = np.linspace(-11, 2.5, 300)
colors = {"Train": "#4f46e5", "Validation": "#06b6d4", "Test": "#10b981"}

for name, ds in zip(split_names, split_datasets):
    kde = gaussian_kde(ds.y.ravel())
    plt.plot(x_grid, kde(x_grid), label=f"{name} (N={len(ds)})", color=colors[name], linewidth=2.0)
    plt.fill_between(x_grid, kde(x_grid), alpha=0.15, color=colors[name])

plt.axvline(-4.0, color="#ef4444", linestyle="--", linewidth=1.3, label="Poor Solubility Threshold (<= -4)")
plt.title("Aqueous Solubility Target Density Across Scaffold Splits", pad=12)
plt.xlabel("Experimental Measured Log Solubility (log mol/L)")
plt.ylabel("Probability Density")
plt.legend(frameon=True, facecolor="white", edgecolor="#cbd5e1", fontsize=9.5)
plt.tight_layout()
plt.show()
```

<h4 style="margin: 0 0 10px 0; color: #1e1b4b; font-size: 16px; font-weight: 700;">
The Role of the Dummy Baseline
</h4>
<p style="font-size: 14px; color: #334155; margin-bottom: 12px;">
Before declaring that any algorithm "understands" chemistry, we must establish what error rate is achievable simply by guessing the training set mean $\bar{y}_{\text{train}}$ for every molecule:
</p>
$$\hat{y}_{\text{dummy}} = \bar{y}_{\text{train}} = -2.867\;\log(\text{mol/L})$$
<p style="font-size: 13.5px; color: #475569; margin: 0;">
Under scaffold splitting, a model with no chemical intelligence yields a negative $R^2$ on the test set due to scaffold distribution shifts. This sets our absolute baseline floor.
</p>
</div>


```python
# Initialize and fit DummyRegressor on training data
dummy_baseline = DummyRegressor(strategy="mean")
dummy_baseline.fit(train_ecfp.X, train_ecfp.y.ravel())

# Predict across splits
pred_dummy_train = dummy_baseline.predict(train_ecfp.X)
pred_dummy_val = dummy_baseline.predict(val_ecfp.X)
pred_dummy_test = dummy_baseline.predict(test_ecfp.X)

def compute_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    return mae, rmse, r2

dummy_eval = {
    "Split": ["Train", "Validation", "Test"],
    "MAE": [
        compute_metrics(train_ecfp.y.ravel(), pred_dummy_train)[0],
        compute_metrics(val_ecfp.y.ravel(), pred_dummy_val)[0],
        compute_metrics(test_ecfp.y.ravel(), pred_dummy_test)[0]
    ],
    "RMSE": [
        compute_metrics(train_ecfp.y.ravel(), pred_dummy_train)[1],
        compute_metrics(val_ecfp.y.ravel(), pred_dummy_val)[1],
        compute_metrics(test_ecfp.y.ravel(), pred_dummy_test)[1]
    ],
    "R² Score": [
        compute_metrics(train_ecfp.y.ravel(), pred_dummy_train)[2],
        compute_metrics(val_ecfp.y.ravel(), pred_dummy_val)[2],
        compute_metrics(test_ecfp.y.ravel(), pred_dummy_test)[2]
    ]
}

print("=" * 70)
print(f"DUMMY REGRESSOR BASELINE (Predicting Mean: {dummy_baseline.constant_[0][0]:.3f})")
print("=" * 70)
display(pd.DataFrame(dummy_eval).round(3))
```

```python
# 1. Train Random Forest Regressor on ECFP Fingerprints
print("=" * 70)
print("TRAINING CANDIDATE 1: RANDOM FOREST REGRESSOR (100 TREES, ECFP)")
print("=" * 70)
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=RANDOM_SEED,
    n_jobs=-1
)
rf_model.fit(train_ecfp.X, train_ecfp.y.ravel())
print(">> Random Forest fitting completed.")

# Predict across splits with Random Forest
pred_rf_train = rf_model.predict(train_ecfp.X)
pred_rf_val = rf_model.predict(val_ecfp.X)
pred_rf_test = rf_model.predict(test_ecfp.X)

# 2. Train DeepChem GraphConvModel on Molecular Graphs
print("=" * 70)
print("TRAINING CANDIDATE 2: DEEPCHEM GRAPHCONVMODEL (GNN, 25 EPOCHS)")
print("=" * 70)
gc_model = GraphConvModel(
    n_tasks=1,
    mode="regression",
    batch_size=64,
    dropout=0.2,
    random_seed=RANDOM_SEED
)

# Train the Graph Neural Network
training_loss = gc_model.fit(train_gc, nb_epoch=25)
print(f">> GraphConvModel fitting completed! Final Loss: {training_loss:.4f}")

# Predict across splits with GraphConvModel
pred_gc_train = gc_model.predict(train_gc).ravel()
pred_gc_val = gc_model.predict(val_gc).ravel()
pred_gc_test = gc_model.predict(test_gc).ravel()
```

<h4 style="margin: 0 0 10px 0; color: #1e1b4b; font-size: 15px; font-weight: 700;">💡 Conceptual Review Questions</h4>
<details style="background: #f8fafc; border: 1px solid #cbd5e1; border-radius: 8px; padding: 12px 16px; margin-bottom: 10px; cursor: pointer;">
<summary style="font-weight: 600; color: #312e81; font-size: 14px;">1. Why is GraphConvModel able to generalize better across scaffolds than Random Forest?</summary>
<p style="font-size: 13.5px; color: #334155; margin-top: 8px; line-height: 1.6;">
Random Forest relies on fixed bit patterns generated by hashing substructures into 1024 bins. If an unseen scaffold produces novel atom combinations whose bits were unpopulated during training, the tree model cannot extrapolate. Graph convolutions pass continuous messages through chemical bonds, allowing the network to learn continuous atomic environments that adapt to novel topologies.
</p>
</details>

<details style="background: #f8fafc; border: 1px solid #cbd5e1; border-radius: 8px; padding: 12px 16px; margin-bottom: 10px; cursor: pointer;">
<summary style="font-weight: 600; color: #312e81; font-size: 14px;">2. What does an MAE of ~1.3 log units mean in real drug discovery?</summary>
<p style="font-size: 13.5px; color: #334155; margin-top: 8px; line-height: 1.6;">
An error of 1.3 log units means predictions are typically within a factor of ~20 of the true molar solubility. While not accurate enough for final formulation design, this level of precision is widely accepted in medicinal chemistry hit-to-lead triage to filter out practically insoluble compounds (log S < -5) from potential orally bioavailable leads.
</p>
</details>

<details style="background: #f8fafc; border: 1px solid #cbd5e1; border-radius: 8px; padding: 12px 16px; margin-bottom: 10px; cursor: pointer;">
<summary style="font-weight: 600; color: #312e81; font-size: 14px;">3. Why did we fit the models without transforming the target units?</summary>
<p style="font-size: 13.5px; color: #334155; margin-top: 8px; line-height: 1.6;">
Log solubility (log S) is already logarithmically transformed by definition, meaning its distribution is approximately normal and homoscedastic. Using unscaled raw units (transformers=[]) ensures that all reported errors (MAE, RMSE) directly reflect physical log mol/L units without requiring back-transformation.
</p>
</details>

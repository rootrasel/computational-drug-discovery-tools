<h4 style="margin: 0 0 10px 0; color: #1e1b4b; font-size: 16px; font-weight: 700;">
Loading Multi-Task Endpoints and the Weight Matrix (w)
</h4>
<p style="font-size: 14px; color: #334155; margin-bottom: 12px;">
Tox21 contains 12 toxicity-related assay endpoints. Each endpoint is a separate binary classification task:
<br>• $y_{i, j} = 1$: Molecule $i$ is active (toxic) in assay $j$.
<br>• $y_{i, j} = 0$: Molecule $i$ is inactive (non-toxic) in assay $j$.
</p>
<div style="background: #ffffff; border: 1px solid #c7d2fe; border-radius: 10px; padding: 14px; margin-bottom: 14px;">
<div style="font-weight: 700; color: #4338ca; font-size: 14px; margin-bottom: 6px;">Understanding the Label Weight Matrix (w):</div>
<div style="font-size: 13.5px; color: #334155; line-height: 1.6;">
In real high-throughput screening, not every molecule was tested across all 12 assays. DeepChem stores an accompanying weight matrix $\mathbf{w}$ of shape $(N, 12)$:
<br>• $w_{i, j} = 1.0$: Molecule $i$ was experimentally measured in assay $j$ (observed label).
<br>• $w_{i, j} = 0.0$: Assay $j$ was not measured for molecule $i$ (missing label).
<br><b>Critical Rule:</b> Unmeasured pairs ($w = 0$) must be excluded from training and evaluation rather than mistakenly treated as negative labels!
</div>
</div>
</div>


```python
# Load the Tox21 benchmark with ECFP featurizer and scaffold splitting
# Redirect stderr briefly to silence featurization of invalid chemical entries
old_stderr = sys.stderr
sys.stderr = io.StringIO()
try:
    tasks, datasets, transformers = dc.molnet.load_tox21(
        featurizer="ECFP",
        splitter="scaffold",
        transformers=[]
    )
finally:
    sys.stderr = old_stderr

train_dataset, valid_dataset, test_dataset = datasets

print("=" * 70)
print("TOX21 BENCHMARK DATASET LOADED")
print("=" * 70)
print("Number of tasks     :", len(tasks))
print("Tasks               :", tasks)
print("Training samples    :", len(train_dataset))
print("Validation samples  :", len(valid_dataset))
print("Test samples        :", len(test_dataset))
print("-" * 70)
print("Feature shape (X)   :", train_dataset.X.shape)
print("Target shape (y)    :", train_dataset.y.shape)
print("Label weights (w)   :", train_dataset.w.shape)
```

Diagnosing Missingness and Class Asymmetry in Tox21
</h4>
<p style="font-size: 14px; color: #334155; margin-bottom: 12px;">
Safety screening datasets in drug discovery typically suffer from <b>extreme class imbalance</b>: the overwhelming majority of compounds tested in a bioassay are inactive (negative, 0), while only a tiny fraction ($3\% - 10\%$) induce toxic disruption (positive, 1).
</p>
<p style="font-size: 13.5px; color: #475569; margin: 0;">
We compute the observed vs. missing counts and active vs. inactive ratio across all 12 tasks in the training set and visualize the distribution.
</p>
</div>


```python
# Explore task labels and class imbalance in training set
label_summary = []
for i, task in enumerate(tasks):
    y = train_dataset.y[:, i]
    w = train_dataset.w[:, i]
    observed = w > 0
    labels = y[observed]
    inactive_cnt = int((labels == 0).sum())
    active_cnt = int((labels == 1).sum())
    total_obs = int(observed.sum())
    pos_rate = (active_cnt / total_obs * 100) if total_obs > 0 else 0
    
    label_summary.append({
        "Task": task,
        "Observed labels": total_obs,
        "Missing labels": int((~observed).sum()),
        "Inactive (0)": inactive_cnt,
        "Active (1)": active_cnt,
        "Active Rate (%)": f"{pos_rate:.1f}%"
    })

label_df = pd.DataFrame(label_summary)
display(label_df)
```

```python
# Plot the positive and negative label counts across all 12 Tox21 assays
fig, ax = plt.subplots(figsize=(14, 5.5), dpi=120)

plot_data = label_df.set_index("Task")[["Inactive (0)", "Active (1)"]]
plot_data.plot(
    kind="bar",
    stacked=False,
    color=["#818cf8", "#ef4444"],
    ax=ax,
    width=0.75,
    edgecolor="white"
)

plt.title("Tox21 Training Label Distribution & Severe Class Imbalance", pad=12)
plt.xlabel("Assay Task", labelpad=10)
plt.ylabel("Number of Observed Molecules", labelpad=10)
plt.xticks(rotation=45, ha="right")
plt.legend(["Inactive (Non-toxic, 0)", "Active (Toxic, 1)"], frameon=True, facecolor="white", edgecolor="#cbd5e1")
plt.tight_layout()
plt.show()
```

<h4 style="margin: 0 0 10px 0; color: #1e1b4b; font-size: 16px; font-weight: 700;">
Independent Binary Classifiers with Class Weight Balancing
</h4>
<p style="font-size: 14px; color: #334155; margin-bottom: 12px;">
To rigorously model Tox21, we train one binary classifier per task:
</p>
<ul style="font-size: 13.5px; color: #334155; margin-left: 20px; line-height: 1.7;">
<li><b>Handling Missing Data:</b> For each task $j$, we filter training and test datasets using the condition $w_j > 0$. This ensures that unmeasured compounds are not misattributed.</li>
<li><b>DummyClassifier Baseline:</b> <code>DummyClassifier(strategy='prior')</code> predicts class probabilities matching the empirical prevalence in the training set, establishing the random-ranking baseline ($\text{ROC-AUC} = 0.50$).</li>
<li><b>Balanced Random Forest:</b> <code>RandomForestClassifier(n_estimators=100, class_weight='balanced')</code> automatically adjusts weights inversely proportional to class frequencies ($w_c = N / (2 \cdot N_c)$) so minority toxic compounds contribute equally during split decisions.</li>
</ul>
</div>


```python
# Helper function to extract observed molecules for a given task
def get_observed_data(dataset, task_idx):
    mask = dataset.w[:, task_idx] > 0
    X = dataset.X[mask]
    y = dataset.y[mask, task_idx].astype(int)
    return X, y

print("=" * 80)
print("TRAINING MULTI-TASK RANDOM FOREST CLASSIFIERS ACROSS ALL 12 TOX21 ASSAYS")
print("=" * 80)

results = []
trained_models = {}

for i, task in enumerate(tasks):
    X_train, y_train = get_observed_data(train_dataset, i)
    X_valid, y_valid = get_observed_data(valid_dataset, i)
    X_test, y_test = get_observed_data(test_dataset, i)

    # Skip tasks without both classes in training or test data
    if len(np.unique(y_train)) < 2 or len(np.unique(y_test)) < 2:
        print(f"Skipping {task}: insufficient class diversity.")
        continue

    # 1. Fit Dummy Baseline Classifier
    baseline = DummyClassifier(strategy="prior")
    baseline.fit(X_train, y_train)

    # 2. Fit Balanced Random Forest Classifier
    rf = RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=RANDOM_SEED,
        n_jobs=-1,
    )
    rf.fit(X_train, y_train)

    # Predictions on Held-Out Scaffold Test Set
    baseline_prob = baseline.predict_proba(X_test)[:, 1]
    rf_prob = rf.predict_proba(X_test)[:, 1]

    # Use standard 0.5 threshold for discrete classification metrics
    rf_pred = (rf_prob >= 0.5).astype(int)

    roc_auc = roc_auc_score(y_test, rf_prob)
    prc_auc = average_precision_score(y_test, rf_prob)
    prec = precision_score(y_test, rf_pred, zero_division=0)
    rec = recall_score(y_test, rf_pred, zero_division=0)
    f1 = f1_score(y_test, rf_pred, zero_division=0)

    results.append({
        "Task": task,
        "Test molecules": len(y_test),
        "Active Positives": int(y_test.sum()),
        "Baseline ROC-AUC": roc_auc_score(y_test, baseline_prob),
        "RF ROC-AUC": roc_auc,
        "RF PRC-AUC": prc_auc,
        "RF Precision": prec,
        "RF Recall": rec,
        "RF F1": f1,
    })
    trained_models[task] = rf

print(">> Successfully trained and evaluated all 12 task classifiers!")
```

```python
# Confusion matrix for highest ROC-AUC task
if trained_models:
    # Select the top-performing task
    best_task = results_df.sort_values("RF ROC-AUC", ascending=False).iloc[0]["Task"]
    task_idx = tasks.index(best_task)
    
    X_test, y_test = get_observed_data(test_dataset, task_idx)
    y_pred = trained_models[best_task].predict(X_test)
    cm = confusion_matrix(y_test, y_pred, labels=[0, 1])

    fig, ax = plt.subplots(figsize=(6, 5), dpi=120)
    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Inactive (0)", "Active (1)"],
    )
    disp.plot(cmap="Blues", ax=ax, colorbar=True)
    ax.set_title(f"Test Confusion Matrix: {best_task}", pad=12)
    ax.grid(False)
    plt.tight_layout()
    plt.show()

    tn, fp, fn, tp = cm.ravel()
    print("=" * 60)
    print(f"CONFUSION MATRIX BREAKDOWN FOR {best_task}")
    print("=" * 60)
    print(f"True Inactive (TN)  : {tn}")
    print(f"False Alarms (FP)   : {fp} (Inactive predicted as Toxic)")
    print(f"Missed Toxins (FN)  : {fn} (Toxic predicted as Inactive)")
    print(f"True Toxins (TP)    : {tp}")
    print(f"Sensitivity (Recall): {tp / (tp + fn) * 100:.1f}%")
    print(f"Specificity         : {tn / (tn + fp) * 100:.1f}%")
```

<div style="font-weight: 700; color: #4338ca; font-size: 13.5px; margin-bottom: 4px;">1. Which Tox21 tasks achieved the highest ROC-AUC?</div>
<div style="font-size: 13px; color: #334155; line-height: 1.6; margin-bottom: 10px;">
<code>NR-AR-LBD</code> (Androgen Receptor Ligand Binding Domain) and <code>NR-AhR</code> (Aryl Hydrocarbon Receptor) achieved the highest discrimination, both surpassing $\text{ROC-AUC} > 0.82$. These nuclear receptors possess well-defined, lipophilic binding pockets that recognize specific steric substructure motifs cleanly captured by Morgan fingerprints.
</div>

<div style="font-weight: 700; color: #4338ca; font-size: 13.5px; margin-bottom: 4px;">2. How did the Random Forest compare with the baseline?</div>
<div style="font-size: 13px; color: #334155; line-height: 1.6; margin-bottom: 10px;">
The DummyClassifier baseline produced a flat $\text{ROC-AUC} = 0.500$ across all 12 tasks (random chance ranking). The balanced Random Forest models significantly outperformed the baseline, achieving a <b>mean ROC-AUC of 0.729</b> across all 12 endpoints.
</div>

<div style="font-weight: 700; color: #4338ca; font-size: 13.5px; margin-bottom: 4px;">3. Which tasks had the strongest class imbalance?</div>
<div style="font-size: 13px; color: #334155; line-height: 1.6; margin-bottom: 10px;">
<code>NR-AR-LBD</code> (only 19 active compounds out of 624 test molecules, $\approx 3.0\%$ active prevalence), <code>NR-ER-LBD</code> ($3.2\%$) and <code>NR-PPAR-gamma</code> ($3.8\%$).
</div>

<div style="font-weight: 700; color: #4338ca; font-size: 13.5px; margin-bottom: 4px;">4. Did high ROC-AUC correspond to good precision and recall?</div>
<div style="font-size: 13px; color: #334155; line-height: 1.6; margin-bottom: 10px;">
Not necessarily. While <code>NR-AR-LBD</code> achieved an impressive $\text{ROC-AUC} = 0.856$, its PRC-AUC was $0.412$ and F1 was $0.467$. In highly imbalanced regimes, even a small false-positive rate from the massive majority class depresses precision. This proves why <b>PRC-AUC and F1-score are vital supplements to ROC-AUC</b> in drug safety evaluation.
</div>

<div style="font-weight: 700; color: #4338ca; font-size: 13.5px; margin-bottom: 4px;">5. How does scaffold-based splitting affect generalization to unseen molecules?</div>
<div style="font-size: 13px; color: #334155; line-height: 1.6;">
By clustering similar Bemis-Murcko scaffolds together, the scaffold split forces the test set to evaluate completely unrepresented core structures. This prevents memorization of chemotypes and realistically simulates prospective safety profiling of novel chemical leads.
</div>
</div>

<h4 style="margin: 18px 0 10px 0; color: #1e1b4b; font-size: 15px; font-weight: 700;">
Limitations of Computational Toxicity Modeling
</h4>
<ul style="font-size: 13px; color: #334155; margin-left: 20px; line-height: 1.7; margin-bottom: 16px;">
<li><b>Missing Assay Labels:</b> Not all compounds are tested in all assays; missingness is not random and can introduce selection bias.</li>
<li><b>Lack of Causal Mechanism:</b> 2D fingerprint models identify statistical correlations between substructures and assay outcomes; they do not elucidate metabolic bioactivation, reactive metabolites, or off-target binding pathways.</li>
<li><b>Experimental Assay Artifacts:</b> Chemical autofluorescence, compound aggregation, and cell cytotoxicity can cause false positives in high-throughput screens.</li>
<li><b>Experimental Requirement:</b> Computational predictions serve as preliminary triage filters to flag liabilities, never as definitive safety clearance.</li>
</ul>

<h4 style="margin: 0 0 10px 0; color: #1e1b4b; font-size: 15px; font-weight: 700;">💡 Conceptual Review Questions</h4>
<details style="background: #f8fafc; border: 1px solid #cbd5e1; border-radius: 8px; padding: 12px 16px; margin-bottom: 10px; cursor: pointer;">
<summary style="font-weight: 600; color: #312e81; font-size: 14px;">1. Why is Accuracy a misleading metric for the Tox21 dataset?</summary>
<p style="font-size: 13.5px; color: #334155; margin-top: 8px; line-height: 1.6;">
Because in assays like NR-AR-LBD, 97% of molecules are inactive. A naive model that classifies 100% of molecules as inactive would achieve 97% accuracy while finding zero toxic compounds. ROC-AUC and PRC-AUC evaluate the model's ranking ability independent of class proportions.
</p>
</details>

<details style="background: #f8fafc; border: 1px solid #cbd5e1; border-radius: 8px; padding: 12px 16px; margin-bottom: 10px; cursor: pointer;">
<summary style="font-weight: 600; color: #312e81; font-size: 14px;">2. How does setting class_weight='balanced' prevent the model from ignoring toxins?</summary>
<p style="font-size: 13.5px; color: #334155; margin-top: 8px; line-height: 1.6;">
Balanced class weighting assigns penalties inversely proportional to class frequencies. Misclassifying an active toxin incurs a penalty ~30x higher than misclassifying an inactive compound, forcing decision trees to split on features that isolate the rare toxic chemotypes.
</p>
</details>

<details style="background: #f8fafc; border: 1px solid #cbd5e1; border-radius: 8px; padding: 12px 16px; margin-bottom: 10px; cursor: pointer;">
<summary style="font-weight: 600; color: #312e81; font-size: 14px;">3. What is the significance of the w array in DeepChem?</summary>
<p style="font-size: 13.5px; color: #334155; margin-top: 8px; line-height: 1.6;">
The w array acts as a binary mask for loss functions and evaluations. In multi-task datasets like Tox21 where assays have unmeasured compounds, w=0 masks out missing entries, allowing models to learn from available data without imputing false zeros.
</p>
</details>

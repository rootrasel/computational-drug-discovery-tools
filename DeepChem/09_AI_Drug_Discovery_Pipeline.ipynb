<h3 style="margin-top: 0; color: #1e1b4b; font-size: 18px; font-weight: 700;">
Target Profile: &beta;-Secretase 1 (BACE-1)
</h3>
<p style="font-size: 14px; color: #334155; margin-bottom: 14px;">
<b>&beta;-Site Amyloid Precursor Protein Cleaving Enzyme 1 (BACE-1)</b> is a transmembrane aspartic protease (peptidase family A1) that performs the initial, rate-limiting proteolytic cleavage of Amyloid Precursor Protein (APP) at the N-terminal Asp672 site. Subsequent intramembrane cleavage by &gamma;-secretase releases pathogenic amyloid-&beta; (A&beta;40 and A&beta;42) peptides whose progressive oligomerization and fibrillar deposition into senile plaques represents a primary neuropathological hallmark of Alzheimer's Disease (AD).
</p>
<ul style="font-size: 13.5px; color: #334155; margin-left: 20px; line-height: 1.7; margin-bottom: 16px;">
<li><b>Active Site Architecture:</b> Catalytic dyad formed by conserved coplanar aspartates (Asp32 and Asp228) operating under general acid-base catalysis.</li>
<li><b>Conformational Dynamics:</b> A flexible 11-residue beta-hairpin "flap" (residues 67&ndash;77) opens to accommodate incoming substrates and closes over bound inhibitors, forming crucial hydrogen bonds (e.g., Gln73, Thr72).</li>
<li><b>Drug Discovery Challenge:</b> Generating small-molecule inhibitors that achieve high target potency, discriminate against related proteases (Cathepsin D), and cross the tight blood-brain barrier (BBB).</li>
</ul>

<h4 style="margin: 0 0 8px 0; color: #1e1b4b; font-size: 15px; font-weight: 700;">
The MoleculeNet BACE Benchmark & Scaffold Splitting
</h4>
<p style="font-size: 14px; color: #334155; margin-bottom: 12px;">
The BACE classification dataset consists of 1,513 curated compounds tested for human BACE-1 enzymatic inhibition. A compound is classified as active (label = 1) if its measured inhibitory concentration satisfies IC<sub>50</sub> &le; 1 &mu;M (pIC<sub>50</sub> &ge; 6.0), and inactive (label = 0) otherwise.
</p>
<div style="background: #f1f5f9; border-left: 4px solid #4f46e5; padding: 12px 16px; border-radius: 6px; font-size: 13.5px; color: #334155; margin-bottom: 12px;">
<b>Why Bemis-Murcko Scaffold Splitting is Mandatory:</b> Random splitting randomly disperses structural analogs across train and test sets, artificially inflating test metrics via "nearest-neighbor memorization." Scaffold splitting groups compounds by their core ring and linker topologies, ensuring that the test set evaluates true out-of-scaffold generalization.
</div>

<p style="margin: 6px 0 0 0; text-align: center; color: #4338ca; font-size: 14px; font-weight: 500;">Training a Balanced Random Forest Classifier on 1024-Bit Circular Fingerprints (ECFP4)</p>

line-height: 1.65;">
<h3 style="margin-top: 0; color: #1e1b4b; font-size: 18px; font-weight: 700;">
Molecular Representation & Classifier Architecture
</h3>
<p style="font-size: 14px; color: #334155; margin-bottom: 14px;">
In Stage 3, small molecules are converted into <b>Extended-Connectivity Fingerprints (ECFP4)</b>, representing circular atom neighborhoods up to a bond radius of 2 (diameter 4) mapped into a 1024-bit binary vector via MurmurHash. ECFP4 fingerprints capture the presence or absence of specific pharmacophoric features (hydrogen-bond donors, aromatic rings, basic nitrogens) critical for protease inhibition.
</p>

<h4 style="margin: 0 0 8px 0; color: #1e1b4b; font-size: 15px; font-weight: 700;">
Predictive Model Training & Probability Calibration
</h4>
<p style="font-size: 14px; color: #334155; margin-bottom: 14px;">
In Stage 4, an ensemble <b>Random Forest Classifier</b> (100 decision trees) is trained on the scaffold-split training partition. To prevent minority class neglect, tree construction employs balanced class weighting:
</p>
<div style="background: #f8fafc; border: 1px solid #cbd5e1; border-radius: 8px; padding: 12px 18px; font-family: monospace; font-size: 13.5px; color: #1e293b; margin-bottom: 14px;">
<i>w</i><sub><i>c</i></sub> = <i>N</i><sub>total</sub> / (2 &times; <i>N</i><sub><i>c</i></sub>)
</div>
<p style="font-size: 14px; color: #334155; margin-bottom: 8px;">
The ensemble outputs a continuous calibrated posterior probability <i>P</i>(Active | <i>x</i>) representing the fraction of decision trees casting an active vote. This probabilistic score enables fine-grained ranking during virtual screening:
</p>
<ul style="font-size: 13.5px; color: #334155; margin-left: 20px; line-height: 1.7; margin-bottom: 0;">
<li><b>High Activity:</b> <i>P</i>(Active) &ge; 0.45 (robust target engagement hypothesis).</li>
<li><b>Moderate Activity:</b> 0.30 &le; <i>P</i>(Active) &lt; 0.45 (intermediate binding likelihood requiring structural confirmation).</li>
<li><b>Low Activity:</b> <i>P</i>(Active) &lt; 0.30 (deprioritized inactive or off-target candidate).</li>
</ul>
</div>


```python
# Initialize and train Balanced Random Forest Classifier
rf_classifier = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_classifier.fit(train_dataset.X, train_dataset.y.ravel())

# Predict probabilities and binary classes
def evaluate_split(X, y_true, split_name):
    y_pred_prob = rf_classifier.predict_proba(X)[:, 1]
    y_pred = (y_pred_prob >= 0.50).astype(int)
    
    roc_auc = roc_auc_score(y_true, y_pred_prob)
    prc_auc = average_precision_score(y_true, y_pred_prob)
    acc = balanced_accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, zero_division=0)
    rec = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)
    
    return {
        "Split": split_name,
        "ROC-AUC": roc_auc,
        "PRC-AUC": prc_auc,
        "Balanced Acc": acc,
        "Precision": prec,
        "Recall": rec,
        "F1-Score": f1,
        "Mean P(Act)": np.mean(y_pred_prob[y_true == 1]),
        "Mean P(Inact)": np.mean(y_pred_prob[y_true == 0])
    }

train_perf = evaluate_split(train_dataset.X, train_dataset.y.ravel(), "Training (1210)")
valid_perf = evaluate_split(valid_dataset.X, valid_dataset.y.ravel(), "Validation (151)")
test_perf = evaluate_split(test_dataset.X, test_dataset.y.ravel(), "Test (152)")

perf_df = pd.DataFrame([train_perf, valid_perf, test_perf])

print("=" * 86)
print("RANDOM FOREST BIOACTIVITY CLASSIFIER EVALUATION (BACE-1 BENCHMARK)")
print("=" * 86)
print(perf_df.to_string(index=False, float_format=lambda x: f"{x:.4f}"))
print("=" * 86)
print(f"\nKey Finding: Test ROC-AUC of {test_perf['ROC-AUC']:.3f} and PRC-AUC of {test_perf['PRC-AUC']:.3f}")
print("demonstrate strong out-of-scaffold discriminatory power on unseen chemical chemotypes.")
```

<h3 style="margin-top: 0; color: #1e1b4b; font-size: 18px; font-weight: 700;">
Physicochemical De-Risking & Early ADMET Screening
</h3>
<p style="font-size: 14px; color: #334155; margin-bottom: 14px;">
Over 40% of promising drug candidates historically failed in clinical development due to unmanageable pharmacokinetic (PK) liabilities or toxicities. In Stage 7, candidate structures are evaluated against established physicochemical boundaries:
</p>

<ul style="font-size: 13.5px; color: #334155; margin-left: 20px; line-height: 1.7; margin-bottom: 14px;">
<li><b>Lipinski's Rule of 5 (Ro5):</b> Molecular Weight (MW) &le; 500 Da, calculated LogP &le; 5.0, Hydrogen Bond Donors (HBD) &le; 5, Hydrogen Bond Acceptors (HBA) &le; 10. Compounds violating &gt; 1 rule typically exhibit poor passive membrane permeability.</li>
<li><b>Veber Bioavailability Parameters:</b> Topological Polar Surface Area (TPSA) &le; 140 &Aring;<sup>2</sup> and Rotatable Bonds &le; 10. For central nervous system (CNS) targets like BACE-1, a tighter TPSA threshold (&le; 90 &Aring;<sup>2</sup>) is preferred for blood-brain barrier penetration.</li>
<li><b>Quantitative Estimate of Drug-likeness (QED):</b> Continuous metric bounded in [0, 1] integrating multiple desirability functions.</li>
<li><b>Structural Alerts (PAINS):</b> Filtering reactive chemical groups (e.g., curcuminoid diketones, Michael acceptors) that cause false-positive assay signals.</li>
</ul>

<h4 style="margin: 0 0 8px 0; color: #1e1b4b; font-size: 15px; font-weight: 700;">
ADMET Risk Classification Rules
</h4>
<table style="width: 100%; border-collapse: collapse; font-size: 13.5px; margin-bottom: 0; text-align: left;">
<thead>
<tr style="background: #eef2ff; color: #1e1b4b; border-bottom: 2px solid #c7d2fe;">
<th style="padding: 8px 12px; font-weight: 700; width: 26%;">ADMET Risk Tier</th>
<th style="padding: 8px 12px; font-weight: 700; width: 34%;">Physicochemical Thresholds</th>
<th style="padding: 8px 12px; font-weight: 700; width: 40%;">Pharmacokinetic Implications</th>
</tr>
</thead>
<tbody>
<tr style="border-bottom: 1px solid #e2e8f0;">
<td style="padding: 8px 12px; font-weight: 700; color: #166534;">Low predicted risk</td>
<td style="padding: 8px 12px;">Ro5 Violations = 0, TPSA &le; 140 &Aring;<sup>2</sup>, RotB &le; 10, no PAINS alerts</td>
<td style="padding: 8px 12px;">Favorable oral bioavailability, adequate intestinal absorption, low metabolic clearance risk.</td>
</tr>
<tr style="border-bottom: 1px solid #e2e8f0; background: #fafafa;">
<td style="padding: 8px 12px; font-weight: 700; color: #b45309;">Moderate predicted risk</td>
<td style="padding: 8px 12px;">Ro5 Violations = 1 OR 140 &lt; TPSA &le; 180 &Aring;<sup>2</sup> OR PAINS warning</td>
<td style="padding: 8px 12px;">Borderline permeability or potential assay interference; warrants chemical optimization.</td>
</tr>
<tr>
<td style="padding: 8px 12px; font-weight: 700; color: #dc2626;">High predicted risk</td>
<td style="padding: 8px 12px;">Ro5 Violations &ge; 2 OR TPSA &gt; 180 &Aring;<sup>2</sup></td>
<td style="padding: 8px 12px;">Severe absorption, bioavailability, or formulation barriers (typical of heavy peptidomimetics).</td>
</tr>
</tbody>
</table>

```python
# Compute RDKit physicochemical descriptors and ADMET risk tiers
for cand, m in zip(screening_cohort, mols):
    mw = Descriptors.MolWt(m)
    logp = Descriptors.MolLogP(m)
    hbd = Descriptors.NumHDonors(m)
    hba = Descriptors.NumHAcceptors(m)
    tpsa = Descriptors.TPSA(m)
    rotb = Descriptors.NumRotatableBonds(m)
    qed_val = QED.qed(m)
    
    # Calculate Lipinski violations
    ro5_violations = sum([
        mw > 500.0,
        logp > 5.0,
        hbd > 5,
        hba > 10
    ])
    
    cand["mw"] = mw
    cand["logp"] = logp
    cand["hbd"] = hbd
    cand["hba"] = hba
    cand["tpsa"] = tpsa
    cand["rotb"] = rotb
    cand["qed"] = qed_val
    cand["ro5_violations"] = ro5_violations
    
    # Identify PAINS / structural alerts
    is_pains = "PAINS" in cand["class"]
    
    # Assign ADMET risk tier
    if ro5_violations == 0 and tpsa <= 140 and rotb <= 10 and not is_pains:
        cand["admet_risk"] = "Low predicted risk"
    elif ro5_violations == 1 or (140 < tpsa <= 180) or is_pains:
        cand["admet_risk"] = "Moderate predicted risk"
    else:
        cand["admet_risk"] = "High predicted risk"

admet_df = pd.DataFrame([
    {
        "Candidate": c["name"],
        "MW (Da)": f"{c['mw']:.1f}",
        "LogP": f"{c['logp']:.2f}",
        "HBD": c["hbd"],
        "HBA": c["hba"],
        "TPSA (Å²)": f"{c['tpsa']:.1f}",
        "RotB": c["rotb"],
        "Ro5 Viol": c["ro5_violations"],
        "QED": f"{c['qed']:.2f}",
        "ADMET Risk": c["admet_risk"]
    }
    for c in screening_cohort
])

print("=" * 96)
print("PHYSICOCHEMICAL DESCRIPTORS & ADMET SAFETY PROFILING")
print("=" * 96)
print(admet_df.to_string(index=False))
print("=" * 96)
```

<h3 style="margin-top: 0; color: #1e1b4b; font-size: 18px; font-weight: 700;">
Multi-Criteria Decision Analysis (MCDA) Framework
</h3>
<p style="font-size: 14px; color: #334155; margin-bottom: 14px;">
Rather than filtering sequentially across isolated silos (which risks discarding valuable chemotypes due to minor artificial cutoffs), modern computational discovery implements <b>Multi-Parameter Optimization (MPO)</b>. The MPO engine concurrently balances predictive bioactivity, docking compatibility, and ADMET risk to classify candidates into actionable decision tiers:
</p>

<table style="width: 100%; border-collapse: collapse; font-size: 13.5px; margin-bottom: 16px; text-align: left;">
<thead>
<tr style="background: #eef2ff; color: #1e1b4b; border-bottom: 2px solid #c7d2fe;">
<th style="padding: 10px 14px; font-weight: 700; width: 18%;">Decision Tier</th>
<th style="padding: 10px 14px; font-weight: 700; width: 44%;">Multi-Criteria Triage Rule</th>
<th style="padding: 10px 14px; font-weight: 700; width: 38%;">Actionable Research Progression</th>
</tr>
</thead>
<tbody>
<tr style="border-bottom: 1px solid #e2e8f0; background: #f0fdf4;">
<td style="padding: 10px 14px; font-weight: 700; color: #166534;">Prioritize</td>
<td style="padding: 10px 14px; color: #166534;">
<b>High/Moderate Activity</b> AND <b>Favorable Docking</b> (&le; &minus;8.5 kcal/mol) AND <b>Low predicted risk</b>
</td>
<td style="padding: 10px 14px; color: #166534;">
Immediate wet-lab procurement/synthesis; priority dispatch to biophysical assays (SPR/ITC) and co-crystallization trials.
</td>
</tr>
<tr style="border-bottom: 1px solid #e2e8f0; background: #fffbeb;">
<td style="padding: 10px 14px; font-weight: 700; color: #92400e;">Review</td>
<td style="padding: 10px 14px; color: #92400e;">
<b>Favorable Docking</b> with <i>Moderate/High ADMET risk</i> (e.g., Peptidomimetics)<br/>
OR <b>Moderate Activity</b> with <i>Moderate Docking / Low Risk</i>
</td>
<td style="padding: 10px 14px; color: #92400e;">
Secondary inspection by medicinal chemistry; explore scaffold modifications (e.g., cyclization, bioisosteric replacement) or formulation strategies.
</td>
</tr>
<tr style="background: #fef2f2;">
<td style="padding: 10px 14px; font-weight: 700; color: #991b1b;">Deprioritize</td>
<td style="padding: 10px 14px; color: #991b1b;">
<b>Low Activity</b> AND <b>Less favorable Docking</b> (&gt; &minus;7.0 kcal/mol) OR off-target control
</td>
<td style="padding: 10px 14px; color: #991b1b;">
Eliminated from further evaluation for this target; archived in internal repository for non-aspartyl protease projects.
</td>
</tr>
</tbody>
</table>
</div>


```python
# Implement the Multi-Parameter Optimization Decision Logic
for cand in screening_cohort:
    act = cand["pred_activity"]
    dock = cand["docking_cat"]
    admet = cand["admet_risk"]
    
    # Prioritize: Strong activity & docking with low ADMET liability
    if act in ["High", "Moderate"] and dock == "Favorable" and admet == "Low predicted risk":
        decision = "Prioritize"
    # Review: Favorable docking (potent pocket fit) that either carries ADMET liabilities (peptidomimetics)
    # or shows low 2D fingerprint similarity (potential ML false negative / novel chemotype)
    elif dock == "Favorable":
        decision = "Review"
    # Review: Moderate activity and moderate docking with clean ADMET
    elif act in ["High", "Moderate"] and dock == "Moderate" and admet in ["Low predicted risk", "Moderate predicted risk"]:
        decision = "Review"
    # Deprioritize: Inactives, decoys, and off-target drugs
    else:
        decision = "Deprioritize"
        
    cand["decision"] = decision

# Format the final screening summary table matching the blueprint specification
screening_summary_df = pd.DataFrame([
    {
        "Candidate": c["name"],
        "Predicted activity": c["pred_activity"],
        "Docking score": c["docking_cat"],
        "ADMET risk": c["admet_risk"],
        "Decision": c["decision"]
    }
    for c in screening_cohort
])

print("=" * 86)
print("FINAL SCREENING SUMMARY TABLE (MULTI-PARAMETER CANDIDATE TRIAGE)")
print("=" * 86)
print(screening_summary_df.to_string(index=False))
print("=" * 86)

# Decision distribution summary
print("\nTriage Decision Distribution:")
dec_dist = screening_summary_df["Decision"].value_counts()
for dec, cnt in dec_dist.items():
    pct = (cnt / len(screening_summary_df)) * 100
    print(f"  {dec:<14}: {cnt:2d} compounds ({pct:5.1f}%)")
```

<h4 style="margin: 0 0 6px 0; color: #1e1b4b; font-size: 14.5px; font-weight: 700;">1. Multi-Criteria Synergy Eliminates False Positives</h4>
<p style="margin: 0; font-size: 13px; color: #475569; line-height: 1.6;">
Neither 2D QSAR nor 3D docking alone is sufficient for robust triage. Donepezil achieves high affinity for AChE but fails BACE-1 docking (&minus;6.8 kcal/mol); classical peptidomimetics (OM99-2) bind tightly (&minus;9.1 kcal/mol) but fail Lipinski bioavailability filters (MW &gt; 650 Da, high TPSA). <h4 style="margin: 0 0 6px 0; color: #1e1b4b; font-size: 14.5px; font-weight: 700;">2. Benchmark Specificity & Applicability Domain</h4>
<p style="margin: 0; font-size: 13px; color: #475569; line-height: 1.6;">
Public benchmarks (MoleculeNet BACE) inherit the chemical space of early patent series (hydroxyethylamines, iminothiadiazines). Highly novel chemotypes with low Tanimoto similarity to the training set produce attenuated probabilities despite real-world activity, underscoring why multi-parameter evidence (combining ML with biophysical docking) is indispensable.
</p>
</div>

<div style="background: #ffffff; border: 1px solid #e2e8f0; border-left: 4px solid #d97706; border-radius: 8px; padding: 14px 16px;">
<h4 style="margin: 0 0 6px 0; color: #1e1b4b; font-size: 14.5px; font-weight: 700;">3. Target Blindness of Ligand-Only Models</h4>
<p style="margin: 0; font-size: 13px; color: #475569; line-height: 1.6;">
Ligand-only models cannot differentiate target isoforms or mutant pockets (e.g., BACE-1 vs. Cathepsin D). Without protein pocket descriptors or structure-based docking poses, a QSAR model cannot predict off-target cross-reactivity or drug resistance mutations.
</p>
</div>

<div style="background: #ffffff; border: 1px solid #e2e8f0; border-left: 4px solid #16a34a; border-radius: 8px; padding: 14px 16px;">
<h4 style="margin: 0 0 6px 0; color: #1e1b4b; font-size: 14.5px; font-weight: 700;">4. The Iterative DMTA Discovery Cycle</h4>
<p style="margin: 0; font-size: 13px; color: #475569; line-height: 1.6;">
Drug discovery is non-linear. Computational screening generates actionable hypotheses (<b>Design</b>), prioritizing compounds for chemical synthesis (<b>Make</b>) and surface plasmon resonance / enzymology assays (<b>Test</b>). The resulting wet-lab experimental bioactivities are fed back to retrain ML models and refine docking scoring functions (<b>Analyze</b>).
</p>
</div>

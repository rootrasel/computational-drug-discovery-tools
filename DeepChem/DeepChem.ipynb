<h2 style="color:#D4AF37;margin-top:0;">
?? 3. Molecular Featurizers
</h2>

<h3 style="color:#4F46E5;">
?? Purpose
</h3>

Molecular featurizers convert chemical structures into numerical representations that machine learning models can understand.

<h3 style="color:#4F46E5;">
?? Common Applications
</h3>

- ? Molecular property prediction
- ? Drug discovery pipelines
- ? QSAR modeling
- ? Graph neural networks

<h3 style="color:#4F46E5;">
?? Key Components
</h3>

- **MolGraphConvFeaturizer**: Converts molecules into graph representations (nodes/edges) for Graph Neural Networks. *(Example: MolGraphConvFeaturizer(use_edges=True))*
- **CircularFingerprint**: Generates Morgan fingerprints, encoding the local environment of each atom into a bit vector. *(Example: CircularFingerprint(size=2048))*
- **ConvMolFeaturizer**: Creates ConvMol objects specifically for DeepChem's GraphConvModel. *(Example: ConvMolFeaturizer())*
- **MACCSKeysFingerprint**: Generates a 166-bit structural fingerprint based on predefined chemical substructures. *(Example: MACCSKeysFingerprint())*
- **RDKitDescriptors**: Computes a large set of standard physicochemical descriptors (e.g. molecular weight, LogP). *(Example: RDKitDescriptors())*

</div>

````python
from deepchem.feat import MolGraphConvFeaturizer
from deepchem.feat import CircularFingerprint
from deepchem.feat import ConvMolFeaturizer
from deepchem.feat import WeaveFeaturizer
from deepchem.feat import MACCSKeysFingerprint
from deepchem.feat import RDKitDescriptors
````

<h2 style="color:#D4AF37;margin-top:0;">
?? 4. Molecule Splitters
</h2>

<h3 style="color:#4F46E5;">
?? Purpose
</h3>

Splitters intelligently divide datasets into training, validation, and test sets to ensure robust and unbiased model evaluation, especially in chemistry where random splits can be misleading.

<h3 style="color:#4F46E5;">
?? Common Applications
</h3>

- ? Scaffold-based splitting for drug discovery
- ? Stratified splitting for imbalanced data
- ? General model evaluation

<h3 style="color:#4F46E5;">
?? Key Components
</h3>

- **RandomSplitter**: Randomly divides datasets. Quick, but can cause data leakage in chemistry. *(Example: RandomSplitter().train_valid_test_split())*
- **ScaffoldSplitter**: Splits based on molecular scaffolds (core structures) ensuring structurally different molecules in the test set. Gold standard for drug discovery. *(Example: ScaffoldSplitter().train_valid_test_split())*
- **RandomStratifiedSplitter**: Ensures the ratio of active/inactive compounds remains identical across splits. Crucial for imbalanced datasets. *(Example: RandomStratifiedSplitter().train_valid_test_split())*

</div>

````python
from deepchem.splits import RandomSplitter
from deepchem.splits import ScaffoldSplitter
from deepchem.splits import IndexSplitter
from deepchem.splits import RandomStratifiedSplitter
````

?? 9. Data Transformers
</h2>

<h3 style="color:#4F46E5;">
?? Purpose
</h3>

Data transformers preprocess datasets before model training by normalizing target values, balancing class distributions, and improving overall model convergence.

<h3 style="color:#4F46E5;">
?? Common Applications
</h3>

- ? Normalize regression targets
- ? Balance imbalanced datasets
- ? Improve model convergence

<h3 style="color:#4F46E5;">
?? Key Components
</h3>

- **NormalizationTransformer**: Scales data to have zero mean and unit variance. Essential for regression tasks. *(Example: NormalizationTransformer(transform_y=True, dataset=dataset))*
- **BalancingTransformer**: Assigns weights to samples to balance out unequal class distributions (e.g., very few toxic compounds). *(Example: BalancingTransformer(dataset=dataset))*

</div>

````python
from deepchem.trans import NormalizationTransformer
from deepchem.trans import BalancingTransformer
````

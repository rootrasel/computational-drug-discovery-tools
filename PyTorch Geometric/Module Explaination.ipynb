1. Graph Representation
</h2>
</div>

### Definition

A graph is a data structure where distinct objects or entities are interconnected. 

In simple terms:
**Graph = Nodes + Edges**

Generally speaking, if we consider an abstract graph:
```text
A ------- B
 \       /
  \     /
     C
```
Here:
* **Nodes:** A, B, C (entities or objects)
* **Edges:** (A-B), (B-C), (A-C) (connections between them)

How does this apply to AI-based Drug Discovery? 

In Drug Discovery, **a molecule is inherently a graph**. 
Where:
* **Atom = Node** 
* **Chemical Bond = Edge**

---

### Real-Life Example (Molecular Graph)

For example, a water (H₂O) molecule:

```text
H — O — H
```

If we represent this as a graph:

```text
Nodes: 
H, O, H

Edges: 
(H, O), (O, H)
```

Remember: an AI model never directly 'reads' chemical formulas like H₂O. It strictly perceives and learns from this topological graph structure.

---

### Drug Discovery Pipeline

Here is the conceptual pipeline illustrating how graphs are utilized in real-world drug discovery:

```text
Chemical Structure
        ↓
Graph Representation
        ↓
Graph Neural Network (GNN)
        ↓
Prediction (Toxicity, Efficacy, etc.)
        ↓
Drug Candidate
```

These are formally known as **Node Features** (or node attributes).

4. Message Passing
</h2>
</div>

This is the most fundamental concept in Graph Neural Networks.

Suppose a carbon atom is bonded to an oxygen atom, which in turn is bonded to a nitrogen atom.

```text
C ---- O ---- N
```

Initially, the carbon atom only contains its own isolated feature vector. However, biochemical properties depend heavily on the local chemical environment. 

Therefore, the oxygen atom transmits contextual information to the carbon atom, and the nitrogen atom transmits information to the oxygen atom.

This iterative exchange and aggregation of localized context is known as **Message Passing**.

---

### Real Life Analogy

Analogous to a clinical team—where doctors, nurses, and patients communicate to update shared care plans—in molecular AI, an atom aggregates information from its neighboring atoms to update its own latent state representation.

5. GCN (Graph Convolutional Network)
</h2>
</div>

What does a GCN do? 
It aggregates and normalizes (averages) information from a node's immediate neighbors.

Example

```text
    H
    |
H — C — O
    |
    H
```

Carbon's neighbors: H, H, H, O

The GCN averages the information across all these neighbors to compute an updated latent representation for the carbon atom. 

---

This enables the AI model to understand the local chemical microenvironment surrounding that carbon atom.

6. GAT (Graph Attention Network)
</h2>
</div>

A standard GCN weights all neighboring nodes equally. In realistic biochemical systems, however, not all neighbors exert equal influence.

Suppose a carbon atom is bonded to:

```text
H
O
N
```

In drug toxicity prediction, an oxygen or halogen neighbor may be vastly more influential than a hydrogen neighbor.

Graph Attention Networks (GATs) dynamically learn these importance scores via self-attention weights.

Example

```text
Hydrogen : 10%
Oxygen   : 70%
Nitrogen : 20%
```

---

Thus, GAT performs adaptive, attention-weighted neighbor aggregation.

---

font-size:30px;">
7. GraphSAGE
</h2>
</div>

In large biomolecular or protein graphs with tens of thousands of atoms, aggregating all neighbors across dense graphs is computationally prohibitive.

GraphSAGE solves this by uniformly or smartly sampling a fixed-size neighborhood.

Example

```text
1000 Neighbor
      ↓
 20 Neighbor
      ↓
  Learning
```

This enables scalable, high-throughput mini-batch training.

8. MPNN (Message Passing Neural Network)
</h2>
</div>

This is the unifying general framework of modern Graph Neural Networks. Most state-of-the-art molecular AI models in drug discovery are built upon the MPNN paradigm.

Step:

```text
Receive Message
      ↓
  Aggregate
      ↓
   Update
      ↓
   Repeat
```

GCN, GAT, and GraphSAGE can all be formulated as specific instances of this generalized MPNN framework.

9. Pooling
</h2>
</div>

Finally, predicting molecular endpoints requires aggregating individual atom embeddings into a unified molecular summary. 

Because predictions are made at the molecular level, graph pooling operations (readout functions) combine all constituent atom features into a single graph-level vector.

```text
200 Atom
   ↓
1 Vector
   ↓
Drug Prediction
```

---

14. Adjacency Matrix
</h2>
</div>

Storing graph connectivity in matrix format.

Suppose a 3-atom system: C, O, N

Connected:
```text
      C O N
C     0 1 0
O     1 0 1
N     0 1 0
```
The adjacency matrix (and its normalized variants) is foundational to spectral and spatial GCN operations.

22. Bipartite Graph
</h2>
</div>

Extensively utilized in Drug Discovery for modeling Drug-Target Interaction (DTI) networks.

```text
Drug
 ↓
 ↓
Protein
```

In a bipartite DTI graph, edges only exist between drugs and target proteins—never between drug-drug or protein-protein pairs.

23. Graph Laplacian
</h2>
</div>

This is the mathematical core of spectral graph theory.

Formula:
```text
L = D − A
```
Where: D = Degree Matrix, A = Adjacency Matrix

In AI Drug Discovery, the Graph Laplacian is utilized for:
* Molecular Similarity
* Spectral Clustering
* Molecular Embedding
* Protein Interaction Analysis
* Graph Signal Processing

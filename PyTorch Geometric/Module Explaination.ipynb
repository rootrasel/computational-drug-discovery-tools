1. Graph Representation
</h2>
</div>

### Definition

Graph হলো এমন একটি data structure যেখানে বিভিন্ন object বা entity একে অপরের সাথে connected থাকে। 

সহজ কথায়:
**Graph = Nodes + Edges**

সাধারণভাবে আমরা যদি চিন্তা করি:
```text
A ------- B
 \       /
  \     /
     C
```
এখানে:
* **Nodes:** A, B, C (যেকোনো object)
* **Edges:** (A-B), (B-C), (A-C) (তাদের মাঝের connection)

কিন্তু AI-based Drug Discovery-তে এটি কীভাবে কাজ করে? 

Drug Discovery-তে, **একটি Molecule-ই হলো একটি Graph**। 
যেখানে:
* **Atom = Node** 
* **Chemical Bond = Edge**

---

### Real-Life Example (Molecular Graph)

ধরি, Water (H₂O) molecule:

```text
H — O — H
```

একে যদি আমরা Graph হিসেবে represent করি, তাহলে:

```text
Nodes: 
H, O, H

Edges: 
(H, O), (O, H)
```

মনে রাখবে, AI কখনো H₂O বা কোনো Chemical Formula লেখা দেখে না। AI শুধু এই Graph Structure-টাই দেখে এবং শেখে।

---

### Drug Discovery Pipeline

বাস্তব জীবনে Drug Discovery-তে Graph কীভাবে ব্যবহৃত হয়, তার একটি Pipeline দেখো:

```text
Chemical Structure
        ↓
Graph Representation
        ↓
Graph Neural Network (GNN)
        ↓
Prediction (Toxicity, Efficacy, etc.)
        ↓
Drug Candidate
```

4. Message Passing
</h2>
</div>

এটাই Graph Neural Network-এর সবচেয়ে গুরুত্বপূর্ণ ধারণা।

ধরো, Carbon-এর পাশে Oxygen আছে, আর Oxygen-এর পাশে Nitrogen আছে।

তাই Oxygen Carbon-কে Information পাঠায়। Nitrogen Oxygen-কে Information পাঠায়।

এই Information Exchange-কে **Message Passing** বলে।

---

### Real Life Analogy

Hospital-এ Doctor, Nurse, Patient সবাই একে অপরের কাছ থেকে Information নিচ্ছে এবং আপডেট হচ্ছে। Drug Discovery-তেও একটি Atom তার Neighbor Atom-এর information সংগ্রহ করে নিজেকে আপডেট করে।

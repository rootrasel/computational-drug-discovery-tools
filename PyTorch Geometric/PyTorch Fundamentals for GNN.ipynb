# Teaching Flow (Drug Discovery Scenario)

Module শুরুতেই student-দের একটা গল্প বলো।

> "আমরা ইতিমধ্যে শিখেছি একটি Molecule হলো একটি Graph।
>
> কিন্তু Graph Neural Network (GNN) তৈরি করার আগে আমাদের জানতে হবে AI কীভাবে সংখ্যা (Tensor) নিয়ে কাজ করে।
>
> কারণ AI কখনো Carbon, Oxygen বা Nitrogen দেখে না।
>
> AI শুধু Tensor দেখে।
>
> তাই PyTorch শেখা হচ্ছে Graph Neural Network শেখার প্রথম ধাপ।"

তারপর একটি ছোট pipeline দেখাতে পারো:

```text
Molecule

↓

RDKit

↓

Graph

↓

Tensor

↓

PyTorch

↓

PyTorch Geometric

↓

Graph Neural Network

↓

Drug Property Prediction
```

এভাবে student-রা বুঝতে পারবে **কেন PyTorch শিখছে**। শুধু API মুখস্থ করবে না, বরং পুরো workflow-এর context বুঝবে।

---

## আমি আরও একটি topic যোগ করার পরামর্শ দেব

Module 2-এ এই দুইটি topic অবশ্যই রাখো:

* **Tensor Operations** (indexing, slicing, reshape, view, transpose)
* **Computational Graph** (Autograd-এর আগে)

কারণ পরে PyTorch Geometric-এ `edge_index`, `node_features`, `batch`, `x`—সবই Tensor হিসেবে থাকবে। Tensor ভালোভাবে না বুঝলে PyG শেখা কঠিন হবে।

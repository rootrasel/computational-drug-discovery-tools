# Teaching Flow (Drug Discovery Scenario)

At the beginning of the module, introduce the topic with an intuitive framing story:

> "We have already learned that a molecule is naturally represented as a graph.
>
> But before building a Graph Neural Network (GNN), we must understand how AI operates on multi-dimensional numbers (Tensors).
>
> Because an AI model never directly sees Carbon, Oxygen, or Nitrogen.
>
> An AI model only perceives Tensors.
>
> Therefore, mastering PyTorch is the essential first step to mastering Graph Neural Networks."

Then, illustrate this conceptual pipeline:

```text
Molecule

↓

RDKit

↓

Graph

↓

Tensor

↓

PyTorch

↓

PyTorch Geometric

↓

Graph Neural Network

↓

Drug Property Prediction
```

This contextual framing helps students understand **why they are learning PyTorch**—moving beyond memorizing APIs to grasp the full end-to-end drug discovery workflow.

---

## Recommended Supplementary Topics

Be sure to include these two foundational topics in Module 2:

* **Tensor Operations** (indexing, slicing, reshape, view, transpose)
* **Computational Graph** (prior to Autograd)

Because in PyTorch Geometric, core data structures like `edge_index`, `node_features`, `batch`, and `x` are all manipulated as tensors. Without a strong grasp of tensor operations, navigating PyG will be challenging.

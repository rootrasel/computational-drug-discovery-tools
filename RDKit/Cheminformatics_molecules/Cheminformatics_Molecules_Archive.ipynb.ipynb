<h2 style="margin: 0; color: #FFFFFF; font-size: 24px; font-weight: 700;">
Bond Types in Cheminformatics
</h2>
<p style="margin: 6px 0 0 0; color: #94A3B8; font-size: 14px;">
Chemical connectivity, bond order definitions, and intermolecular interaction types in computational representations
</p>
</div>

In chemical computer files (such as MDL Molfile, SDF, or SMILES strings), chemical bonds are primarily categorized by the following numerical orders or symbolic notations:

* **Single Bond:** Encoded in connection tables as bond order `1`, or denoted by `-` in SMILES (typically implicit, e.g., `CC`).
* **Double Bond:** Encoded in connection tables as bond order `2`, or denoted by `=` in SMILES (e.g., `C=C`).
* **Triple Bond:** Encoded in connection tables as bond order `3`, or denoted by `#` in SMILES (e.g., `C#N`).
* **Aromatic Bond:** Represents delocalized pi-electron systems (e.g., benzene); encoded in connection tables as bond order `4`, represented by `:` in explicit SMILES or designated via lower-case atomic symbols (e.g., `c1ccccc1`).
* **Amide Bond:** A specialized carbonyl-nitrogen single bond with partial double-bond character, fundamental in peptidomimetics, protein backbones, and drug design.
* **Coordinate / Dative Bond:** A dipolar bond where both shared electrons originate from a single donor atom, frequently found in metal complexes and coordination chemistry.
* **Hydrogen Bond:** A directional non-covalent dipole interaction essential for modeling target-ligand binding affinities and conformational stability.

<h2 style="margin: 0; color: #FFFFFF; font-size: 24px; font-weight: 700;">
Compound Categories in Cheminformatics
</h2>
<p style="margin: 6px 0 0 0; color: #94A3B8; font-size: 14px;">
Chemical taxonomy, database curation filters, and molecular structural classes
</p>
</div>

In cheminformatics databases (such as PubChem, ChEBI, and ChEMBL), chemical substances are primarily categorized and filtered into the following core classes:

* **Organic Compounds:** Carbon- and hydrogen-based chemical entities, constituting the primary foundation of small-molecule drug design.
* **Inorganic Compounds:** Mineral- or metal-based compounds lacking carbon-hydrogen frameworks.
* **Organometallic Compounds:** Compounds featuring direct chemical bonds between carbon atoms and metallic elements.
* **Macrocycles / Polymers:** Large-ring molecules or repeating-chain high-molecular-weight chemical entities.
* **Biomolecules / Macromolecules:** Naturally occurring biopolymers including proteins, DNA, RNA, lipids, and carbohydrates.
* **Small Molecules:** Low molecular weight chemical entities (typically MW < 900 Daltons), representing the vast majority of therapeutic drugs.
* **Salts and Hydrates:** Multi-component crystal forms consisting of ionized parent drugs paired with counterions or stoichiometric water molecules.
* **Mixtures / Ratios:** Formulations composed of two or more distinct chemical entities in defined stoichiometric proportions.

In cheminformatics pipelines and data curation workflows, compound datasets are curated, standardized, and filtered across these fundamental chemical categories.

# 1️⃣ Bond Types in Cheminformatics

কম্পিউটার ফাইলে (যেমন: Molfile বা SMILES-এ) বন্ডগুলোকে মূলত নিচের সংখ্যা বা সাইন দ্বারা ক্যাটাগরি করা হয়:

* **Single Bond (একক বন্ধন):** ফাইলে এটিকে `1` বা `-` সাইন দিয়ে প্রকাশ করা হয়।
* **Double Bond (দ্বিবন্ধন):** ফাইলে এটিকে `2` বা `=` সাইন দিয়ে প্রকাশ করা হয়।
* **Triple Bond (ত্রিবন্ধন):** ফাইলে এটিকে `3` বা `#` সাইন দিয়ে প্রকাশ করা হয়।
* **Aromatic Bond (অ্যারোমেটিক বন্ধন):** বেনজিন রিংয়ের মতো চক্রাকার বন্ধন, ফাইলে এটিকে `4` বা `:` অথবা ছোট হাতের অক্ষর (যেমন: বেনজিনের জন্য `c`) দিয়ে প্রকাশ করা হয়।
* **Amide Bond (অ্যামাইড বন্ধন):** পেপটাইড এবং প্রোটিন চেইনের জন্য বিশেষ বন্ধন টাইপ।
* **Coordinate/Dative Bond (সন্নিবেশ বন্ধন):** মেটাল-অর্গানিক যৌগের জন্য ইলেকট্রন জোড় শেয়ারিং বন্ধন।
* **Hydrogen Bond (হাইড্রোজেন বন্ধন):** অণুগুলোর মধ্যকার আকর্ষণ বল (Intermolecular forces) গণনার জন্য ব্যবহৃত হয়।

# 2️⃣ Compound Categories in Cheminformatics

Cheminformatics Database (such as: PubChem, ChEBI, ChEMBL) যৌগগুলোকে প্রধানত এই কয়েকটি ভাগে ফিল্টার বা ক্যাটাগরি করা থাকে:

* **Organic Compounds (অর্গানিক/জৈব যৌগ):** কার্বন ও হাইড্রোজেন ভিত্তিক যৌগ (ড্রাগ ডিজাইনের মূল উপাদান)।
* **Inorganic Compounds (ইনঅর্গানিক/অজৈব যৌগ):** কার্বন ছাড়া অন্যান্য খনিজ বা ধাতু ভিত্তিক যৌগ।
* **Organometallic Compounds (অর্গানোমেটালিক যৌগ):** কার্বনের সাথে সরাসরি ধাতুর বন্ধনযুক্ত যৌগ।
* **Macrocycles / Polymers (ম্যাক্রোসাইকেল ও পলিমার):** বড় আকৃতির পুনরাবৃত্তিমূলক চেইনের যৌগ।
* **Biomolecules / Macromolecules (জৈব অণু):** প্রোটিন, ডিএনএ (DNA), আরএনএ (RNA) এবং কার্বোহাইড্রেট।
* **Small Molecules (ছোট অণু):** কম আণবিক ওজনের যৌগ (Molecular weight < 900 Daltons), যা মূলত ওষুধ তৈরিতে ব্যবহৃত হয়।
* **Salts and Hydrates (লবণ ও হাইড্রেট):** আয়নিত অবস্থায় থাকা যৌগ বা ক্রিস্টাল ফর্মে পানির অণুযুক্ত যৌগ।
* **Mixtures / Ratios (মিশ্রণ):** একাধিক যৌগের নির্দিষ্ট অনুপাতের কম্বিনেশন।

কেমিনফরম্যাটিক্সের সম্পূর্ণ সিলেবাস বা ডেটা অ্যানালাইসিসে যৌগ এবং বন্ধনের এই ফরম্যাট ও ক্যাটাগরিগুলোর ভেতরেই সবকিছু সীমাবদ্ধ থাকে।

# 🧩 Complete SMILES & SMARTS Symbol Cheat Sheet

This is a comprehensive reference guide for molecular string representations, summarizing elements from earlier sections.

### 1. Atom Symbols
| Symbol | Meaning | Example |
|---|---|---|
| `C`, `c` | Carbon, Aromatic Carbon | `CC`, `c1ccccc1` |
| `N`, `n` | Nitrogen, Aromatic Nitrogen | `CCN`, `n1ccccc1` |
| `O`, `o` | Oxygen, Aromatic Oxygen | `CCO`, `o1cccc1` |
| `S`, `s` | Sulfur, Aromatic Sulfur | `CS`, `s1cccc1` |
| `P`, `p` | Phosphorus, Aromatic Phosphorus | `CP`, *(Rare)* |
| `F`, `Cl`, `Br`, `I` | Halogens | `CF`, `CCl`, `CBr`, `CI` |
| `H` | Hydrogen (explicit) | `[H]O` |
| `B`, `Si`, `Se` | Boron, Silicon, Selenium | `B(O)O`, `[Si]`, `[Se]` |
| Metals | Mg, Na, K, Ca, Fe, Cu, Zn | `[Mg+2]`, `[Na+]`, `[Fe]` |

### 2. Bond & Connectivity Symbols
| Symbol | Meaning | Example |
|---|---|---|
| *(implicit)* | Single Bond | `CC` |
| `-` | Single Bond | `C-C` |
| `=` | Double Bond | `C=C` |
| `#` | Triple Bond | `C#N` |
| `:` | Aromatic Bond | `c:c` |
| `.` | Disconnected Molecules / Ionic Pair | `[Na+].[Cl-]` |
| `->`, `<-` | Coordinate Bond | `[NH3->Cu]` |
| `1`–`9`, `%10` | Ring Closures | `C1CCCCC1`, `%10CCCCC%10` |
| `()` | Branching | `CC(C)O` |

### 3. Charge, Isotope & Stereochemistry
| Symbol | Meaning | Example |
|---|---|---|
| `+`, `-`, `+2`, `-2`| Charge | `[NH4+]`, `[Cl-]`, `[Ca+2]` |
| `@`, `@@` | Chiral Center (Tetrahedral) | `C[C@H](O)N`, `C[C@@H](O)N` |
| `/`, `\` | E/Z Double Bond | `F/C=C/F`, `F/C=C\F` |
| `[13C]`, `[2H]` | Isotopes | `[13CH4]`, `[2H]O` |
| `[]` | Explicit Atom brackets | `[Na+]`, `[H]` |

### 4. SMARTS Wildcards & Logic
| Symbol | Meaning | Symbol | Meaning |
|---|---|---|---|
| `*` | Any Atom | `!` | NOT |
| `A`, `a` | Any Aliphatic/Aromatic Atom | `,` | OR |
| `R`, `r` | Ring Atom, Ring Size | `;`, `&`| AND (Low/High precedence) |
| `X`, `x` | Connectivity, Ring Connectivity | `$()` | Recursive SMARTS |
| `D` | Degree | `>>` | Reaction Arrow (`CCO>>CC=O`) |

### 5. Common Functional Group SMARTS Patterns
| Group | SMARTS | Group | SMARTS |
|---|---|---|---|
| Alcohol | `[OX2H]` | Ketone | `C(=O)C` |
| Amine | `[NX3]` | Aldehyde | `[CH]=O` |
| Carboxylic Acid| `C(=O)O` | Ether | `COC` |
| Ester | `C(=O)O[C]` | Nitrile | `C#N` |
| Amide | `C(=O)N` | Nitro | `[N+](=O)[O-]` |

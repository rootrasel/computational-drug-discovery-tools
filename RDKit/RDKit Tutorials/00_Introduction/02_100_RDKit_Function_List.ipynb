### What is SMILES?
- SMILES stands for Simplified Molecular Input Line Entry System.

### Why use SMILES?
- Computers cannot easily understand pictures of molecules, but they can easily process text.

| Molecule | Structure (simplified)   | SMILES     |
| -------- | ------------------------ | ---------- |
| Water    | H–O–H                    | `O`        |
| Methane  | CH₄                      | `C`        |
| Ethane   | C–C                      | `CC`       |
| Ethanol  | C–C–O                    | `CCO`      |
| Propane  | C–C–C                    | `CCC`      |
| Ethene   | C=C                      | `C=C`      |
| Benzene  | Six-carbon aromatic ring | `c1ccccc1` |


Instead of drawing a chemical structure like this:

```
    H   H
    |   |
H - C - C - O - H
    |   |
    H   H
```
we write it as a simple string:

CCO

Both represent ethanol (alcohol).

# Reading Simple SMILES (With Examples)

## 1. Carbon is `C`

A single `C` represents one carbon atom.

**Example:**
| Molecule | SMILES | Structure |
|----------|---------|-----------|
| Methane | `C` | CH₄ |



## 2. Two Carbons

`CC` means two carbon atoms connected by a single bond.

**Example:**
| Molecule | SMILES | Structure |
|----------|---------|-----------|
| Ethane | `CC` | C—C |



## 3. Three Carbons

`CCC` represents three carbon atoms in a chain.

**Example:**
| Molecule | SMILES | Structure |
|----------|---------|-----------|
| Propane | `CCC` | C—C—C |



## 4. Oxygen is `O`

`CCO` means a carbon chain ending with an oxygen atom.

**Example:**
| Molecule | SMILES | Structure |
|----------|---------|-----------|
| Ethanol | `CCO` | C—C—O |



## 5. Nitrogen is `N`

`CCN` represents a carbon chain attached to nitrogen.

**Example:**
| Molecule | SMILES | Structure |
|----------|---------|-----------|
| Ethylamine | `CCN` | C—C—N |



# Bonds in SMILES

## Single Bond

SMILES:
```text
CC
```

Structure:
```text
C—C
```

**Example:** Ethane



## Double Bond

SMILES:
```text
C=C
```

Structure:
```text
C═C
```

**Example:** Ethene



## Triple Bond

SMILES:
```text
C#N
```

Structure:
```text
C≡N
```

**Example:** Hydrogen cyanide (HCN)



# Rings

Numbers indicate where a ring starts and ends.

SMILES:
```text
c1ccccc1
```

Represents:

```text
      C
   /     \
  C       C
  ||     ||
  C       C
   \     /
      C
```

**Example:** Benzene (six-membered aromatic ring)



# Branches

Parentheses `()` indicate side chains (branches).

SMILES:
```text
CC(C)O
```

Structure:

```text
      C
      |
C — C — O
```

**Example:** Isopropanol (2-Propanol)

Another branching example:

SMILES:
```text
CC(C)C
```

Structure:

```text
      C
      |
C — C — C
```

**Example:** Isobutane (2-Methylpropane)

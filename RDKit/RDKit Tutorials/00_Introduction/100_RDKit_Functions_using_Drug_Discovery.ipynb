````python

# Module 1 — Reading & Writing Molecules
from rdkit import Chem
from rdkit.Chem import AllChem


# 1. Chem.MolFromSmiles()
smiles = "CCO"   # Ethanol

mol = Chem.MolFromSmiles(smiles)

print("-"*60)
print("1. MolFromSmiles()")
print("-"*60)
print("Input SMILES :", smiles)
print("RDKit Mol    :", mol)



# 2. Chem.MolToSmiles()
canonical_smiles = Chem.MolToSmiles(mol)

print("\n" + "-"*60)
print("2. MolToSmiles()")
print("-"*60)
print("Canonical SMILES :", canonical_smiles)



# Generate 2D Coordinates
AllChem.Compute2DCoords(mol)



# 3. Chem.MolToMolBlock()
molblock = Chem.MolToMolBlock(mol)

print("\n" + "-"*60)
print("3. MolToMolBlock()")
print("-"*60)
print(molblock[:400])     # Print first few lines only



# 4. Chem.MolFromMolBlock()
mol_from_block = Chem.MolFromMolBlock(molblock)

print("\n" + "-"*60)
print("4. MolFromMolBlock()")
print("-"*60)
print("Recovered SMILES :", Chem.MolToSmiles(mol_from_block))



# 5. Chem.MolToMolFile()
Chem.MolToMolFile(
    mol,
    "ethanol.mol"
)

print("\nSaved : ethanol.mol")



# 6. Chem.MolFromMolFile()
mol_from_file = Chem.MolFromMolFile("ethanol.mol")

print("\n" + "-"*60)
print("6. MolFromMolFile()")
print("-"*60)
print("Recovered SMILES :", Chem.MolToSmiles(mol_from_file))



# 7. Chem.SDWriter()
writer = Chem.SDWriter("ethanol.sdf")
writer.write(mol)
writer.close()

print("\nSaved : ethanol.sdf")



# 8. Chem.SDMolSupplier()
supplier = Chem.SDMolSupplier("ethanol.sdf")

print("\n" + "-"*60)
print("8. SDMolSupplier()")
print("-"*60)

for molecule in supplier:
    if molecule is not None:
        print("Loaded :", Chem.MolToSmiles(molecule))



# 9. Chem.MolToInchi()
inchi = Chem.MolToInchi(mol)

print("\n" + "-"*60)
print("9. MolToInchi()")
print("-"*60)
print(inchi)



# 10. Chem.MolFromInchi()
mol_from_inchi = Chem.MolFromInchi(inchi)

print("\n" + "-"*60)
print("10. MolFromInchi()")
print("-"*60)
print("Recovered SMILES :", Chem.MolToSmiles(mol_from_inchi))

print("\n" + "-"*60)
print("Module 1 Completed Successfully!")
print("-"*60)
````

````python
from rdkit import Chem
from rdkit.Chem import AllChem


for name, smiles in molecules.items():

    print(f"Molecule : {name}")
    print(f"SMILES   : {smiles}")

    mol = Chem.MolFromSmiles(smiles)

    # -------------------------------------------------
    # 1. Number of atoms
    # -------------------------------------------------
    print("\n1. Number of Atoms")
    print(mol.GetNumAtoms())

    # -------------------------------------------------
    # 2. Number of bonds
    # -------------------------------------------------
    print("\n2. Number of Bonds")
    print(mol.GetNumBonds())

    # -------------------------------------------------
    # 3. GetAtoms()
    # -------------------------------------------------
    print("\n3. Atoms")

    for atom in mol.GetAtoms():
        print(
            f"Index={atom.GetIdx():2d} "
            f"Symbol={atom.GetSymbol():2s} "
            f"AtomicNumber={atom.GetAtomicNum():2d}"
        )

    # -------------------------------------------------
    # 4. GetBonds()
    # -------------------------------------------------
    print("\n4. Bonds")

    for bond in mol.GetBonds():
        print(
            f"{bond.GetBeginAtomIdx()} --> {bond.GetEndAtomIdx()}   "
            f"{bond.GetBondType()}"
        )

    # -------------------------------------------------
    # 5. GetAtomWithIdx()
    # -------------------------------------------------
    print("\n5. Atom With Index")

    if mol.GetNumAtoms() > 0:
        atom = mol.GetAtomWithIdx(0)

        print("Atom 0")
        print("Symbol :", atom.GetSymbol())
        print("Degree :", atom.GetDegree())
        print("Charge :", atom.GetFormalCharge())

    # -------------------------------------------------
    # 6. GetBondWithIdx()
    # -------------------------------------------------
    print("\n6. Bond With Index")

    if mol.GetNumBonds() > 0:

        bond = mol.GetBondWithIdx(0)

        print("Bond 0")
        print("Begin Atom :", bond.GetBeginAtomIdx())
        print("End Atom   :", bond.GetEndAtomIdx())
        print("Bond Type  :", bond.GetBondType())

    # -------------------------------------------------
    # 7. Ring Information
    # -------------------------------------------------
    print("\n7. Ring Information")

    ring = mol.GetRingInfo()

    print("Number of Rings :", ring.NumRings())

    # -------------------------------------------------
    # 8. Conformer (3D Coordinates)
    # -------------------------------------------------
    print("\n8. Conformer")

    try:

        mol3d = Chem.AddHs(mol)

        AllChem.EmbedMolecule(mol3d)

        conf = mol3d.GetConformer()

        pos = conf.GetAtomPosition(0)

        print("Atom0 Coordinates")
        print(f"x={pos.x:.3f}")
        print(f"y={pos.y:.3f}")
        print(f"z={pos.z:.3f}")

    except:
        print("Could not generate conformer")

    # -------------------------------------------------
    # 9. Formal Charges
    # -------------------------------------------------
    print("\n9. Formal Charges")

    for atom in mol.GetAtoms():

        print(
            atom.GetSymbol(),
            "Charge =",
            atom.GetFormalCharge()
        )

    # -------------------------------------------------
    # 10. Molecular Fragments
    # -------------------------------------------------
    print("\n10. Molecular Fragments")

    frags = Chem.GetMolFrags(mol)

    print(frags)
````

Module 3 — Molecular Descriptors (10)
- `Descriptors.MolWt()` - Calculates the average molecular weight of a molecule.
- `Descriptors.ExactMolWt()` - Calculates the exact molecular weight using the exact isotopic masses of atoms.
- `Crippen.MolLogP()` - Measures the lipophilicity (oil vs. water solubility) of a molecule.
- `Descriptors.TPSA()` - Calculates the Topological Polar Surface Area (TPSA), indicating molecular polarity and membrane permeability.
- `Lipinski.NumHAcceptors()` - Counts the number of Hydrogen Bond Acceptors (HBA) in a molecule.
- `Lipinski.NumHDonors()` - Counts the number of Hydrogen Bond Donors (HBD) in a molecule.
- `Lipinski.NumRotatableBonds()` - Counts the number of rotatable bonds, indicating molecular flexibility.
- `Descriptors.HeavyAtomCount()` - Counts all non-hydrogen (heavy) atoms in a molecule.
- `Descriptors.RingCount()` - Counts the total number of rings present in a molecule.
- `Descriptors.FractionCSP3()` - Calculates the fraction of sp³-hybridized carbon atoms, indicating the molecule's three-dimensional character.


````python
from rdkit.Chem import Descriptors
from rdkit.Chem import Crippen
from rdkit.Chem import Lipinski


# Select a Molecule
molecule_name = "Ethanol"

mol = rdkit_molecules[molecule_name]
smiles = molecules[molecule_name]


# Molecular Descriptors
mol_weight = Descriptors.MolWt(mol)
exact_weight = Descriptors.ExactMolWt(mol)
logp = Crippen.MolLogP(mol)
tpsa = Descriptors.TPSA(mol)
h_acceptors = Lipinski.NumHAcceptors(mol)
h_donors = Lipinski.NumHDonors(mol)
rotatable_bonds = Lipinski.NumRotatableBonds(mol)
heavy_atoms = Descriptors.HeavyAtomCount(mol)
ring_count = Descriptors.RingCount(mol)
fraction_csp3 = Descriptors.FractionCSP3(mol)


# Print Results
print("=" * 60)
print(f"Molecule Name           : {molecule_name}")
print(f"SMILES                  : {smiles}")
print("=" * 60)

print(f"Average Molecular Weight : {mol_weight:.2f}")
print(f"Exact Molecular Weight   : {exact_weight:.4f}")
print(f"LogP                    : {logp:.2f}")
print(f"TPSA                    : {tpsa:.2f}")
print(f"H-Bond Acceptors        : {h_acceptors}")
print(f"H-Bond Donors           : {h_donors}")
print(f"Rotatable Bonds         : {rotatable_bonds}")
print(f"Heavy Atom Count        : {heavy_atoms}")
print(f"Ring Count              : {ring_count}")
print(f"Fraction Csp3           : {fraction_csp3:.2f}")

print("=" * 60)
````

# Module 4 — Molecular Fingerprints

### 1. `AllChem.GetMorganFingerprintAsBitVect()`
**Purpose:** Generates a Morgan (ECFP) circular fingerprint, widely used for molecular similarity, QSAR, and virtual screening.



### 2. `MACCSkeys.GenMACCSKeys()`
**Purpose:** Generates a 166-bit MACCS structural fingerprint based on predefined chemical substructure keys.



### 3. `Chem.RDKFingerprint()`
**Purpose:** Generates RDKit's default path-based fingerprint using atom paths within the molecule.



### 4. `Pairs.GetAtomPairFingerprint()`
**Purpose:** Generates an Atom Pair fingerprint based on atom types and the topological distance between them.



### 5. `Torsions.GetTopologicalTorsionFingerprint()`
**Purpose:** Generates a Topological Torsion fingerprint using sequences of connected atoms (torsions).



### 6. `DataStructs.TanimotoSimilarity()`
**Purpose:** Calculates the Tanimoto similarity between two molecular fingerprints.



### 7. `DataStructs.DiceSimilarity()`
**Purpose:** Calculates the Dice similarity coefficient between two molecular fingerprints.



### 8. `DataStructs.CosineSimilarity()`
**Purpose:** Calculates the Cosine similarity between two molecular fingerprints.



### 9. `BulkTanimotoSimilarity()`
**Purpose:** Computes Tanimoto similarity between one fingerprint and multiple fingerprints efficiently.



### 10. `ConvertToNumpyArray()`
**Purpose:** Converts an RDKit fingerprint into a NumPy array for machine learning and data analysis.

````python
import numpy as np

from rdkit import Chem, DataStructs
from rdkit.Chem import AllChem
from rdkit.Chem import MACCSkeys
from rdkit.Chem.AtomPairs import Pairs
from rdkit.Chem.AtomPairs import Torsions


# Select Two Molecules
molecule_1 = "Aspirin"
molecule_2 = "Ibuprofen"

mol1 = rdkit_molecules[molecule_1]
mol2 = rdkit_molecules[molecule_2]

print("="*70)
print(f"Molecule 1 : {molecule_1}")
print(f"SMILES     : {molecules[molecule_1]}")
print()

print(f"Molecule 2 : {molecule_2}")
print(f"SMILES     : {molecules[molecule_2]}")
print("="*70)


# 1. Morgan Fingerprint (ECFP)
morgan1 = AllChem.GetMorganFingerprintAsBitVect(mol1, radius=2, nBits=2048)
morgan2 = AllChem.GetMorganFingerprintAsBitVect(mol2, radius=2, nBits=2048)


# 2. MACCS Fingerprint
maccs1 = MACCSkeys.GenMACCSKeys(mol1)
maccs2 = MACCSkeys.GenMACCSKeys(mol2)


# 3. RDKit Fingerprint
rdk1 = Chem.RDKFingerprint(mol1)
rdk2 = Chem.RDKFingerprint(mol2)


# 4. Atom Pair Fingerprint
pair1 = Pairs.GetAtomPairFingerprint(mol1)
pair2 = Pairs.GetAtomPairFingerprint(mol2)


# 5. Topological Torsion Fingerprint
torsion1 = Torsions.GetTopologicalTorsionFingerprint(mol1)
torsion2 = Torsions.GetTopologicalTorsionFingerprint(mol2)


# 6. Tanimoto Similarity
tanimoto = DataStructs.TanimotoSimilarity(morgan1, morgan2)


# 7. Dice Similarity
dice = DataStructs.DiceSimilarity(morgan1, morgan2)


# 8. Cosine Similarity
cosine = DataStructs.CosineSimilarity(morgan1, morgan2)


# 9. Bulk Tanimoto Similarity
bulk = DataStructs.BulkTanimotoSimilarity(
    morgan1,
    [morgan2]
)


# 10. Convert Fingerprint to NumPy Array
array = np.zeros((2048,), dtype=int)

DataStructs.ConvertToNumpyArray(morgan1, array)


# Print Results
print("\nSimilarity Scores")
print("-"*70)

print(f"Tanimoto Similarity : {tanimoto:.3f}")
print(f"Dice Similarity     : {dice:.3f}")
print(f"Cosine Similarity   : {cosine:.3f}")
print(f"Bulk Tanimoto       : {bulk}")

print("\nFingerprint Information")
print("-"*70)

print(f"Morgan Bits     : {morgan1.GetNumBits()}")
print(f"MACCS Bits      : {maccs1.GetNumBits()}")
print(f"RDKit Bits      : {rdk1.GetNumBits()}")

print("\nNumPy Array")
print("-"*70)

print("Shape :", array.shape)
print("First 64 Bits:")

print(array[:64])

print("="*70)
````

````python
# Module 6 — Substructure Search
from rdkit import Chem
from rdkit.Chem import Fragments
from rdkit.Chem import Draw
from rdkit.Chem import rdFMCS
from IPython.display import display
````

````python

# Example Molecule
smiles = "CC(=O)OC1=CC=CC=C1C(=O)O"   # Aspirin
mol = Chem.MolFromSmiles(smiles)

display(Draw.MolToImage(mol, size=(350,350)))


# 1. Chem.MolFromSmarts()
print("-"*60)
print("1. MolFromSmarts()")
print("-"*60)

benzene = Chem.MolFromSmarts("c1ccccc1")

print("SMARTS Pattern: c1ccccc1")


# 2. HasSubstructMatch()
print("\n" + "-"*60)
print("2. HasSubstructMatch()")
print("-"*60)

print(mol.HasSubstructMatch(benzene))


# 3. GetSubstructMatch()
print("\n" + "-"*60)
print("3. GetSubstructMatch()")
print("-"*60)

match = mol.GetSubstructMatch(benzene)

print(match)

img = Draw.MolToImage(
    mol,
    highlightAtoms=list(match),
    size=(400,400)
)

display(img)


# 4. GetSubstructMatches()
print("\n" + "-"*60)
print("4. GetSubstructMatches()")
print("-"*60)

matches = mol.GetSubstructMatches(benzene)

print(matches)


# 5. DeleteSubstructs()
print("\n" + "-"*60)
print("5. DeleteSubstructs()")
print("-"*60)

deleted = Chem.DeleteSubstructs(
    mol,
    benzene
)

display(Draw.MolToImage(deleted))


# 6. ReplaceSubstructs()
print("\n" + "-"*60)
print("6. ReplaceSubstructs()")
print("-"*60)

fluorine = Chem.MolFromSmiles("F")

replaced = Chem.ReplaceSubstructs(
    mol,
    benzene,
    fluorine
)

display(Draw.MolToImage(replaced[0]))


# 7. Fragments.fr_amide()
print("\n" + "-"*60)
print("7. fr_amide()")
print("-"*60)

print("Amide Count :", Fragments.fr_amide(mol))


# 8. Fragments.fr_ester()
print("\n" + "-"*60)
print("8. fr_ester()")
print("-"*60)

print("Ester Count :", Fragments.fr_ester(mol))


# 9. Fragments.fr_benzene()
print("\n" + "-"*60)
print("9. fr_benzene()")
print("-"*60)

print("Benzene Rings :", Fragments.fr_benzene(mol))


# 10. rdFMCS.FindMCS()
print("\n" + "-"*60)
print("10. FindMCS()")
print("-"*60)

mol2 = Chem.MolFromSmiles("Oc1ccccc1")   # Phenol

mcs = rdFMCS.FindMCS([mol, mol2])

print("SMARTS :", mcs.smartsString)
print("Atoms  :", mcs.numAtoms)
print("Bonds  :", mcs.numBonds)

mcs_pattern = Chem.MolFromSmarts(mcs.smartsString)

match1 = mol.GetSubstructMatch(mcs_pattern)
match2 = mol2.GetSubstructMatch(mcs_pattern)

grid = Draw.MolsToGridImage(
    [mol, mol2],
    legends=["Aspirin", "Phenol"],
    highlightAtomLists=[
        list(match1),
        list(match2)
    ],
    molsPerRow=2,
    subImgSize=(350,350)
)

display(grid)

print("\n" + "-"*60)
print("Module 6 Completed Successfully!")
print("-"*60)
````

````python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem.Scaffolds import MurckoScaffold
from rdkit.Chem import BRICS
from rdkit.Chem import Recap
from IPython.display import display
````

````python
# Example Molecule
smiles = "CC(=O)OC1=CC=CC=C1C(=O)O"      # Aspirin
mol = Chem.MolFromSmiles(smiles)

print("-"*60)
print("Original Molecule")
print("-"*60)

display(Draw.MolToImage(mol, size=(350,350)))


# 1. MurckoScaffold.GetScaffoldForMol()
print("\n" + "-"*60)
print("1. Murcko Scaffold")
print("-"*60)

scaffold = MurckoScaffold.GetScaffoldForMol(mol)

display(Draw.MolToImage(scaffold, size=(300,300)))


# 2. MurckoScaffold.MakeScaffoldGeneric()
print("\n" + "-"*60)
print("2. Generic Scaffold")
print("-"*60)

generic = MurckoScaffold.MakeScaffoldGeneric(scaffold)

display(Draw.MolToImage(generic, size=(300,300)))


# 3. BRICS.BRICSDecompose()
print("\n" + "-"*60)
print("3. BRICS Decomposition")
print("-"*60)

brics_fragments = BRICS.BRICSDecompose(mol)

for fragment in brics_fragments:
    print(fragment)


# 4. BRICS.BreakBRICSBonds()
print("\n" + "-"*60)
print("4. Break BRICS Bonds")
print("-"*60)

broken = BRICS.BreakBRICSBonds(mol)

display(Draw.MolToImage(broken))

# 5. Recap.RecapDecompose()
print("\n" + "-"*60)
print("5. RECAP Decomposition")
print("-"*60)

recap_tree = Recap.RecapDecompose(mol)

print(recap_tree.children.keys())


# 6. Recap.GetLeaves()
print("\n" + "-"*60)
print("6. RECAP Leaves")
print("-"*60)

leaves = recap_tree.GetLeaves()

for leaf in leaves:
    print(leaf)


# 7. Chem.FragmentOnBonds()
print("\n" + "-"*60)
print("7. Fragment On Bonds")
print("-"*60)

fragmented = Chem.FragmentOnBonds(
    mol,
    bondIndices=[0]
)

display(Draw.MolToImage(fragmented))


# 8. Chem.GetMolFrags()
print("\n" + "-"*60)
print("8. GetMolFrags")
print("-"*60)

fragments = Chem.GetMolFrags(
    fragmented,
    asMols=True
)

for i, frag in enumerate(fragments):

    print(f"Fragment {i+1}")

    display(Draw.MolToImage(frag))


# 9. Chem.CombineMols()
print("\n" + "-"*60)
print("9. Combine Molecules")
print("-"*60)

benzene = Chem.MolFromSmiles("c1ccccc1")
ethanol = Chem.MolFromSmiles("CCO")

combined = Chem.CombineMols(
    benzene,
    ethanol
)

display(Draw.MolToImage(combined))


# 10. EditableMol()
print("\n" + "-"*60)
print("10. EditableMol")
print("-"*60)

editable = Chem.EditableMol(benzene)

oxygen = Chem.Atom("O")

oxygen_idx = editable.AddAtom(oxygen)

editable.AddBond(
    0,
    oxygen_idx,
    Chem.BondType.SINGLE
)

new_mol = editable.GetMol()

display(Draw.MolToImage(new_mol))

print("\n" + "-"*60)
print("Module 7 Completed Successfully!")
print("-"*60)
````

````python
# Example Molecule
smiles = "CC(C)CC1=CC=C(C=C1)C(C)C(=O)O"   # Ibuprofen

mol = Chem.MolFromSmiles(smiles)

print("-"*60)
print("Original Molecule")
print("-"*60)

print("Atoms:", mol.GetNumAtoms())


# 1. Chem.AddHs()
print("\n" + "-"*60)
print("1. AddHs()")
print("-"*60)

molH = Chem.AddHs(mol)

print("Atoms Before :", mol.GetNumAtoms())
print("Atoms After  :", molH.GetNumAtoms())


# 2. Chem.RemoveHs()
print("\n" + "-"*60)
print("2. RemoveHs()")
print("-"*60)

mol_noH = Chem.RemoveHs(molH)

print("Atoms After Removing H :", mol_noH.GetNumAtoms())


# 3. EmbedMolecule()
print("\n" + "-"*60)
print("3. EmbedMolecule()")
print("-"*60)

status = AllChem.EmbedMolecule(molH)

print("Embedding Status :", status)


# 4. EmbedMultipleConfs()
print("\n" + "-"*60)
print("4. EmbedMultipleConfs()")
print("-"*60)

conf_ids = AllChem.EmbedMultipleConfs(
    molH,
    numConfs=5
)

print("Conformer IDs :", list(conf_ids))


# 5. UFFOptimizeMolecule()
print("\n" + "-"*60)
print("5. UFFOptimizeMolecule()")
print("-"*60)

result = AllChem.UFFOptimizeMolecule(molH)

print("Optimization Status :", result)


# 6. MMFFOptimizeMolecule()
print("\n" + "-"*60)
print("6. MMFFOptimizeMolecule()")
print("-"*60)

result = AllChem.MMFFOptimizeMolecule(molH)

print("Optimization Status :", result)


# 7. MMFFGetMoleculeProperties()
print("\n" + "-"*60)
print("7. MMFFGetMoleculeProperties()")
print("-"*60)

properties = AllChem.MMFFGetMoleculeProperties(molH)

print(properties)


# 8. UFFGetMoleculeForceField()
print("\n" + "-"*60)
print("8. UFFGetMoleculeForceField()")
print("-"*60)

forcefield = AllChem.UFFGetMoleculeForceField(
    molH
)

energy = forcefield.CalcEnergy()

print("UFF Energy :", energy)


# 9. ComputeGasteigerCharges()
print("\n" + "-"*60)
print("9. ComputeGasteigerCharges()")
print("-"*60)

AllChem.ComputeGasteigerCharges(molH)

for atom in molH.GetAtoms():

    charge = atom.GetProp("_GasteigerCharge")

    print(
        f"Atom {atom.GetIdx():2d} "
        f"{atom.GetSymbol():2s} "
        f"Charge = {float(charge): .4f}"
    )


# 10. CanonicalRankAtoms()
print("\n" + "-"*60)
print("10. CanonicalRankAtoms()")
print("-"*60)

ranks = Chem.CanonicalRankAtoms(mol)

for atom, rank in zip(mol.GetAtoms(), ranks):

    print(
        f"Atom {atom.GetIdx():2d} "
        f"{atom.GetSymbol():2s} "
        f"Rank = {rank}"
    )
````

| `CalcMolFormula()`                       | Calculate the molecular formula (e.g., C₉H₈O₄).                           |
| `CalcNumRings()`                         | Count the number of rings in a molecule.                                  |
| `CalcChi0v()`                            | Calculate the Chi0v connectivity index (a topological descriptor).        |
| `CalcHallKierAlpha()`                    | Compute the Hall–Kier alpha value, reflecting atomic environment effects. |
| `CalcLabuteASA()`                        | Calculate the approximate Labute Accessible Surface Area (ASA).           |
| `CalcAUTOCORR2D()`                       | Generate 2D autocorrelation descriptors for machine learning features.    |


````python
import pandas as pd

from rdkit import Chem
from rdkit.Chem import PandasTools
from rdkit.Chem import Descriptors
from rdkit.Chem import rdMolDescriptors


# Example Dataset
df = pd.DataFrame({

    "Name":[
        "Water",
        "Ethanol",
        "Benzene",
        "Aspirin",
        "Caffeine"
    ],

    "SMILES":[
        "O",
        "CCO",
        "c1ccccc1",
        "CC(=O)OC1=CC=CC=C1C(=O)O",
        "Cn1cnc2n(C)c(=O)n(C)c(=O)c12"
    ]

})

print("-"*60)
print("Original Dataset")
print("-"*60)

display(df)


# 1. PandasTools.LoadSDF()
print("\n" + "-"*60)
print("1. LoadSDF()")
print("-"*60)

print("Example:")
print("""
# sdf = PandasTools.LoadSDF(
#     'chembl.sdf',
#     smilesName='SMILES',
#     molColName='ROMol'
# )
""")


# 2. AddMoleculeColumnToFrame()
print("\n" + "-"*60)
print("2. AddMoleculeColumnToFrame()")
print("-"*60)

PandasTools.AddMoleculeColumnToFrame(
    df,
    smilesCol="SMILES",
    molCol="Mol"
)

display(df)


# 3. SaveSMILESFromFrame()
print("\n" + "-"*60)
print("3. SaveSMILESFromFrame()")
print("-"*60)

PandasTools.SaveSMILESFromFrame(
    df,
    "molecules.smi",
    molCol="Mol",
    NamesCol="Name"
)

print("Saved : molecules.smi")


# 4. Descriptors._descList
print("\n" + "-"*60)
print("4. Descriptor List")
print("-"*60)

print("Total Descriptors :", len(Descriptors._descList))

print("\nFirst 10 Descriptors")

for name, func in Descriptors._descList[:10]:

    print(name)


# 5. CalcMolFormula()
print("\n" + "-"*60)
print("5. Molecular Formula")
print("-"*60)

df["Formula"] = df["Mol"].apply(
    rdMolDescriptors.CalcMolFormula
)

display(df[["Name","Formula"]])


# 6. CalcNumRings()
print("\n" + "-"*60)
print("6. Number of Rings")
print("-"*60)

df["NumRings"] = df["Mol"].apply(
    rdMolDescriptors.CalcNumRings
)

display(df[["Name","NumRings"]])


# 7. CalcChi0v()
print("\n" + "-"*60)
print("7. Chi0v")
print("-"*60)

df["Chi0v"] = df["Mol"].apply(
    rdMolDescriptors.CalcChi0v
)

display(df[["Name","Chi0v"]])


# 8. CalcHallKierAlpha()
print("\n" + "-"*60)
print("8. Hall Kier Alpha")
print("-"*60)

df["HallKierAlpha"] = df["Mol"].apply(
    rdMolDescriptors.CalcHallKierAlpha
)

display(df[["Name","HallKierAlpha"]])


# 9. CalcLabuteASA()
print("\n" + "-"*60)
print("9. Labute ASA")
print("-"*60)

df["LabuteASA"] = df["Mol"].apply(
    rdMolDescriptors.CalcLabuteASA
)

display(df[["Name","LabuteASA"]])


# 10. CalcAUTOCORR2D()
print("\n" + "-"*60)
print("10. AUTOCORR2D")
print("-"*60)

autocorr = rdMolDescriptors.CalcAUTOCORR2D(df["Mol"][3])

print("Length :", len(autocorr))

print("\nFirst 10 Values")

print(autocorr[:10])

print("\n" + "-"*60)
print("Module 10 Completed Successfully!")
print("-"*60)
````

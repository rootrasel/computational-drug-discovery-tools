## Classification of Molecular Descriptors (Cheminformatics)

**Molecular descriptors** are numerical representations of molecules that convert chemical structures into mathematical values. They are used in **QSAR (Quantitative Structure–Activity Relationship)**, **drug discovery**, **machine learning**, and **molecular property prediction**.

They are mainly classified based on the dimensionality of molecular information:

There are five types of Molecular Descriptors
- 1. 0D Molecular Descriptors (Constitutional Descriptors)
- 2. 1D Molecular Descriptors (Structural Fragment Descriptors)
- 3. 2D Molecular Descriptors (Topological Descriptors)
- 4. 3D Molecular Descriptors (Spatial Descriptors)
- 5. 4D Molecular Descriptors (Dynamic Descriptors)

---

## 1. 0D Molecular Descriptors (Constitutional Descriptors)

**Definition:**
Descriptors that use only the **molecular formula and atom counts**. They do not contain information about atom connectivity or 3D structure.

### Information included:

* Molecular formula
* Number of atoms
* Number of heavy atoms
* Molecular weight
* Element composition
* Atom charges
* Electronegativity
* Polarizability

### Examples:

For Aspirin (C₉H₈O₄):

* Molecular weight = 180.16 g/mol
* Number of carbon atoms = 9
* Number of oxygen atoms = 4
* Hydrogen bond acceptors = 4

### Applications:

* Basic molecular property prediction
* Filtering molecules in databases

---

# 2. 1D Molecular Descriptors (Structural Fragment Descriptors)

**Definition:**
Descriptors based on **functional groups and substructures** present in molecules.

### Information included:

* Functional groups
* Chemical fragments
* Substituents
* Structural patterns

### Examples:

* Alcohol group (-OH)
* Carboxylic acid (-COOH)
* Aromatic rings
* Amide groups

### Fingerprints (Extremely Important)

A special type of 1D molecular descriptor. Fingerprints are the heart of similarity search and AI-driven drug discovery.

### What They Are

A fingerprint represents a molecule's fragments as a binary vector.

**Example — Aspirin:**
```
Aspirin

010010101011010101010
```


Examples:

* Morgan fingerprints (ECFP)
* MACCS keys

They represent molecular fragments as binary vectors:

A molecule becomes a binary vector::



Where:
* 1 → fragment exists
* 0 → fragment absent

### Applications:

* Similarity searching
* Virtual screening
* Drug discovery

- `1` → fragment exists
- `0` → fragment absent

### Applications

- Similarity searching
- Virtual screening
- Drug discovery

### Types of Fingerprints

**A. Morgan Fingerprint (Most Important)**
- Also called: ECFP (Extended Connectivity Fingerprint)
- Used in:
  - Similarity search
  - ML models
  - Virtual screening

**B. MACCS Fingerprint**
- 166 predefined structural keys

**C. Atom Pair Fingerprint**
- Uses atom distance information

**D. Topological Torsion Fingerprint**
- Uses atom sequences

---

# 3. 2D Molecular Descriptors (Topological Descriptors)

**Definition:**
Descriptors based on **molecular connectivity and graph structure**.

A molecule is represented as:

```
Atoms → Nodes
Bonds → Edges
```

They ignore the actual 3D arrangement.

### Information included:

* Atom connectivity
* Bond types
* Molecular graph

### Examples:

### a) Topological indices:

* Wiener index
* Balaban index
* Zagreb index

### b) Connectivity descriptors:

* Degree of branching
* Path-based descriptors

### c) SMILES-based descriptors:

Example:

```
Aspirin SMILES:

CC(=O)OC1=CC=CC=C1C(=O)O
```

### Applications:

* QSAR modeling
* ADMET prediction
* Machine learning models

---

# 4. 3D Molecular Descriptors (Spatial Descriptors)

**Definition:**
Descriptors that use the **three-dimensional coordinates of atoms**.

They require molecular geometry.

Example:

```
Atom coordinates:

C1  x=0.23 y=1.45 z=2.11
O1  x=1.22 y=0.34 z=0.55
```

### Information included:

* Atom positions
* Molecular shape
* Distances between atoms
* Surface properties

### Examples:

* Molecular volume
* Solvent accessible surface area (SASA)
* 3D pharmacophore descriptors
* Shape descriptors

### Applications:

* Protein-ligand interaction
* Docking studies
* Structure-based drug design

---

# 5. 4D Molecular Descriptors (Dynamic Descriptors)

**Definition:**
Descriptors that include **time-dependent molecular behavior**.

They extend 3D descriptors by considering molecular flexibility and movement.

### Information included:

* Multiple conformations
* Molecular dynamics
* Protein-ligand interactions
* Conformational changes

### Examples:

* Average molecular shape over time
* Dynamic pharmacophore
* Molecular flexibility descriptors

### Applications:

* Advanced drug discovery
* Protein-ligand modeling

---

# Summary Table

| Dimension | Main Information                | Examples                                 |
| --------- | ------------------------------- | ---------------------------------------- |
| **0D**    | Formula and atom counts         | MW, atom number, HBD, HBA                |
| **1D**    | Functional groups and fragments | Morgan fingerprint, MACCS                |
| **2D**    | Molecular graph connectivity    | Topological indices, SMILES descriptors  |
| **3D**    | Spatial arrangement             | Volume, shape, surface area              |
| **4D**    | Dynamic molecular behavior      | Molecular motion, conformational changes |

---

### In modern AI drug discovery:

Typical pipeline:

```
SMILES
  |
  ↓
RDKit
  |
  ↓
Molecular Descriptors
  |
  ↓
Machine Learning Model
  |
  ↓
Activity / Toxicity / ADMET Prediction
```

For example using RDKit:

```python
from rdkit import Chem
from rdkit.Chem import Descriptors

mol = Chem.MolFromSmiles(
    "CC(=O)OC1=CC=CC=C1C(=O)O"
)

print(Descriptors.MolWt(mol))
print(Descriptors.NumHAcceptors(mol))
print(Descriptors.NumHDonors(mol))
```

Output:

```
180.16
4
1
```

These descriptors become the **features (X)** that machine learning models use for molecular prediction.

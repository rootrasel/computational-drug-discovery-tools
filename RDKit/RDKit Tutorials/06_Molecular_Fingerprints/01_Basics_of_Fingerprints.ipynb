A **Molecular Fingerprint** is a digital vector representation that encodes the 2D or 3D chemical structure of a molecule into a **fixed-length binary bit string (or count vector)**. Because machine learning models and database indexing systems cannot parse chemical sketches directly, fingerprints provide a standardized numerical abstraction of structural features.

In a molecular fingerprint, each bit represents the presence ($1$) or absence ($0$) of a designated structural fragment, circular neighborhood, or chemical pattern.

Molecular fingerprints are widely utilized in **cheminformatics, drug discovery, and AI-driven screening**. Key applications include:
* **Molecular Similarity Searching:** Rapidly identifying structural analogs of active compounds across libraries containing millions of chemicals.
* **Virtual Screening:** Triaging massive chemical catalogs to enrich for potential lead drug candidates.
* **Machine Learning Feature Engineering:** Supplying robust input features for predicting biological activity, toxicity, solubility, and target binding.

### Key Fingerprint Types in RDKit

1. **Morgan Fingerprints (ECFP — Extended Connectivity Fingerprints):**
   The industry gold standard for medicinal chemistry and QSAR. Encodes circular atomic neighborhoods up to a designated radius (e.g., radius 2 corresponds to ECFP4) into fixed-length vectors (typically 1024 or 2048 bits).

2. **MACCS Keys (166 Bits):**
   A predefined dictionary of 166 structural motifs and functional groups, offering high interpretability and computational efficiency.

3. **Atom Pair Fingerprints:**
   Encodes pairs of atoms along with their topological graph distance.

4. **Topological Torsion Fingerprints:**
   Encodes four consecutively bonded non-hydrogen atoms, capturing conformational and flexibility characteristics.

In an AI drug discovery pipeline, molecules are ingested as SMILES, converted into RDKit Mol objects, featurized into Morgan bit vectors, and supplied to machine learning models (Random Forest, XGBoost, Neural Networks) to predict binding affinities and ADMET properties.

> **Molecular Fingerprints serve as the vital computational bridge between chemical structures and predictive AI models, enabling automated screening, structural comparison, and potency forecasting.**

<h2 style="margin: 0; color: #FFFFFF; font-size: 20px; font-weight: 700;">
1. Morgan Fingerprint (ECFP) — Circular Topology
</h2>
</div>

## Concept

Morgan fingerprint represents a molecule by looking at the environment around every atom.

Example:

```
       O
       |
C ---- C ---- C
       |
       H
```

RDKit expands the environment:

```
Radius 0:
Atom itself

Radius 1:
Atom + directly connected atoms

Radius 2:
Atom + neighbors + neighbors of neighbors
```

Then converts these environments into bits:

```
Molecule

       ↓

Circular fragments

       ↓

Hashing

       ↓

2048-bit vector


[0,1,0,0,1,1,0,0,1.....]
```

---

## RDKit Example

```python
from rdkit import Chem
from rdkit.Chem import AllChem

mol = Chem.MolFromSmiles(
    "CC(=O)OC1=CC=CC=C1C(=O)O"
)

fp = AllChem.GetMorganFingerprintAsBitVect(
    mol,
    radius=2,
    nBits=2048
)

print(fp)
```

Parameters:

| Parameter  | Meaning     |
| ---------- | ----------- |
| radius=2   | ECFP4       |
| radius=3   | ECFP6       |
| nBits=2048 | Vector size |

<h2 style="margin: 0; color: #FFFFFF; font-size: 20px; font-weight: 700;">
2. MACCS Keys (166 Predefined Structural Patterns)
</h2>
</div>

## Concept

MACCS uses predefined chemical rules.

Example:

```
Does molecule contain:

☑ Aromatic ring
☑ Alcohol group
☑ Amide group
☑ Halogen
☑ Carbonyl
```

Output:

```
[1,0,1,1,0,0,1....]

166 bits
```

---

## RDKit

```python
from rdkit.Chem import MACCSkeys

fp = MACCSkeys.GenMACCSKeys(mol)
```

---

## Advantage

Very interpretable:

Example:

Bit 162:

```
Aromatic ring present = Yes
```

Morgan:

```
Bit 1057 = unknown fragment
```

MACCS:

```
Bit 1057 does not exist
```

<h2 style="margin: 0; color: #FFFFFF; font-size: 20px; font-weight: 700;">
3. Atom Pair Fingerprints (Topological Distance Vectors)
</h2>
</div>

## Concept

Instead of local environment, it stores:

```
Atom type
+
Distance between atoms
```

Example:

Molecule:

```
C ---- C ---- O
```

Stored information:

```
Carbon - Oxygen

Distance = 2 bonds
```

Representation:

```
(C,O,2)
```

---

## Why useful?

Morgan:

```
Looks locally
```

Atom Pair:

```
Looks globally
```

Example:

Two molecules:

```
A:

C-C-C-O


B:

C-C-C-O

```

Even if skeleton changes slightly, distance information remains useful.

---

## RDKit

```python
from rdkit.Chem import rdMolDescriptors

fp = rdMolDescriptors.GetHashedAtomPairFingerprintAsBitVect(
    mol
)
```

4. Topological Torsion Fingerprints (Four-Atom Paths)
</h2>
</div>

## Concept

Looks at sequences of 4 connected atoms.

Example:

```
C - C - N - O
```

Stored as:

```
C-C-N-O
```

It captures:

* Backbone pattern
* Flexible chains
* Conformation-related information

---

## RDKit

```python
from rdkit.Chem import AllChem

fp = AllChem.GetHashedTopologicalTorsionFingerprintAsBitVect(
    mol
)
```

# Notebook Structure

## Module 1 — Import Libraries

````python
import pandas as pd

from rdkit import Chem
from rdkit.Chem import Descriptors
from rdkit.Chem import Lipinski
````

---

## Module 2 — Load Dataset

````python
dataset_path = "../../dataset/molecules_100.csv"

df = pd.read_csv(dataset_path)

df.head()
````

---

## Module 3 — Convert SMILES

````python
df["Mol"] = df["SMILES"].apply(Chem.MolFromSmiles)

df.head()
````

---

## Module 4 — Calculate Molecular Weight

````python
df["MolecularWeight"] = df["Mol"].apply(
    Descriptors.MolWt
)

df.head()
````

---

## Module 5 — Calculate LogP

````python
df["LogP"] = df["Mol"].apply(
    Descriptors.MolLogP
)

df.head()
````

---

## Module 6 — Calculate Hydrogen Bond Donors

````python
df["HBD"] = df["Mol"].apply(
    Lipinski.NumHDonors
)

df.head()
````

---

## Module 7 — Calculate Hydrogen Bond Acceptors

````python
df["HBA"] = df["Mol"].apply(
    Lipinski.NumHAcceptors
)

df.head()
````

---

# Module 8 — Apply Lipinski Rule

````python
def lipinski_rule(mol):

    mw = Descriptors.MolWt(mol)

    logp = Descriptors.MolLogP(mol)

    hbd = Lipinski.NumHDonors(mol)

    hba = Lipinski.NumHAcceptors(mol)

    violations = 0

    if mw > 500:
        violations += 1

    if logp > 5:
        violations += 1

    if hbd > 5:
        violations += 1

    if hba > 10:
        violations += 1

    return violations
````

---

## Module 9 — Count Violations

````python
df["Violations"] = df["Mol"].apply(
    lipinski_rule
)

df.head()
````

---

## Module 10 — Drug-like Classification

````python
df["DrugLike"] = df["Violations"].apply(
    lambda x: "Yes" if x <= 1 else "No"
)

df.head()
````

---

## Module 11 — Final Report

````python
result = df[
    [
        "Name",
        "MolecularWeight",
        "LogP",
        "HBD",
        "HBA",
        "Violations",
        "DrugLike"
    ]
]

result
````

---

## Module 12 — Drug-like Molecules Only

````python
drug_like = result[
    result["DrugLike"] == "Yes"
]

drug_like
````

## Module 13 — Non Drug-like Molecules

````python
non_drug = result[
    result["DrugLike"] == "No"
]

non_drug
````

---

## Module 14 — Summary Statistics

````python
print("="*50)

print("Total Molecules :", len(df))

print("Drug-like :", len(drug_like))

print("Non Drug-like :", len(non_drug))

print("="*50)
````
